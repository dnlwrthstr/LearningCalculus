## 1. Introduction

We know that the derivative of a function $f(x)$ is defined by the limit

$$
f'(x)=\frac{d}{dx}f(x)=\lim_{\Delta x \to 0}\frac{f(x+\Delta x)-f(x)}{\Delta x}
$$

This definition is fundamental, but using it every time is slow and algebraically heavy.

For polynomial functions, there are special rules that make differentiation much faster.

In this notebook, we will study:
- the derivative of a constant
- the derivative of $x$
- the power rule
- derivatives of polynomial expressions
- the product rule
- the quotient rule

By the end, you should be able to differentiate many polynomial expressions quickly and correctly.

## 2. Theory

### 2.1 Derivative of a constant

Consider the constant function

$$
f(x)=c
$$

Since the value of the function does not change, its increment is always zero:

$$
f(x+\Delta x)-f(x)=c-c=0
$$

Therefore,

$$
f'(x)=\lim_{\Delta x\to 0}\frac{0}{\Delta x}=0
$$

So the rule is:

$$
(c)'=0
$$

---

### 2.2 Derivative of $x$

Now consider

$$
f(x)=x
$$

Then

$$
f(x+\Delta x)=x+\Delta x
$$

So

$$
f'(x)=\lim_{\Delta x\to 0}\frac{(x+\Delta x)-x}{\Delta x}
=\lim_{\Delta x\to 0}\frac{\Delta x}{\Delta x}=1
$$

Thus,

$$
(x)'=1
$$

## 3. The Power Rule

Let us look at

$$
f(x)=x^2
$$

Then

$$
f(x+\Delta x)=(x+\Delta x)^2=x^2+2x\Delta x+(\Delta x)^2
$$

So

$$
\frac{f(x+\Delta x)-f(x)}{\Delta x}
=
\frac{x^2+2x\Delta x+(\Delta x)^2-x^2}{\Delta x}
=
\frac{2x\Delta x+(\Delta x)^2}{\Delta x}
=
2x+\Delta x
$$

Taking the limit as $\Delta x \to 0$ gives

$$
(x^2)'=2x
$$

Similarly,

$$
(x^3)'=3x^2
$$

This leads to the **power rule**:

$$
(x^n)' = n x^{n-1}, \qquad n>0
$$

This is one of the most important rules in calculus.

## 4. Constant Multiple Rule

If $c$ is a constant and $f(x)$ is a differentiable function, then

$$
(cf)' = c f'
$$

Combining this with the power rule gives

$$
(cx^n)' = c \cdot n x^{n-1}
$$

For example,

$$
(5x^4)' = 5 \cdot 4 x^3 = 20x^3
$$

## 5. Sum and Difference Rules

For differentiable functions $f$ and $g$:

$$
(f+g)' = f' + g'
$$

and

$$
(f-g)' = f' - g'
$$

This means we can differentiate a polynomial term by term.

For example, if

$$
f(x)=3x^3+5x^2+8x+1
$$

then

$$
f'(x)=(3x^3)' + (5x^2)' + (8x)' + (1)'
$$

$$
f'(x)=9x^2 + 10x + 8
$$

## 6. Product Rule

The derivative of a product is **not** just the product of the derivatives.

The correct rule is:

$$
(fg)' = f'g + fg'
$$

Example:

$$
(x^2)' = (x\cdot x)' = x'\cdot x + x\cdot x' = 1\cdot x + x\cdot 1 = 2x
$$

Another example:

$$
(3x^4 \cdot 5x^2)' = (3x^4)'\cdot 5x^2 + 3x^4\cdot (5x^2)'
$$

$$
=12x^3 \cdot 5x^2 + 3x^4 \cdot 10x
$$

$$
=60x^5 + 30x^5 = 90x^5
$$

## 7. Quotient Rule

For a quotient of differentiable functions:

$$
\left(\frac{f}{g}\right)' = \frac{f'g - fg'}{g^2}
$$

provided $g(x) \neq 0$.

Example:

$$
f(x)=\frac{x^5+2}{x+4}
$$

Then

$$
f'(x)=\frac{(x^5+2)'(x+4) - (x^5+2)(x+4)'}{(x+4)^2}
$$

$$
=\frac{5x^4(x+4) - (x^5+2)(1)}{(x+4)^2}
$$

Expanding the numerator:

$$
5x^4(x+4) - (x^5+2)=5x^5+20x^4-x^5-2
$$

$$
=4x^5+20x^4-2
$$

So

$$
f'(x)=\frac{4x^5+20x^4-2}{(x+4)^2}
$$

This corrected simplification is important: the constant term is **$-2$**, not $+2$.

## 8. Python Example: Symbolic Differentiation with SymPy

We now verify several rules symbolically.

```
x = sp.symbols('x')

expressions = [
    3,
    x,
    x**2,
    x**3,
    5*x**4,
    3*x**3 + 5*x**2 + 8*x + 1,
    (3*x**4) * (5*x**2),
    (x**5 + 2) / (x + 4)
]

for expr in expressions:
    print("f(x) =", expr)
    print("f'(x) =", sp.simplify(sp.diff(expr, x)))
    print("-" * 50)
```

## 9. Visualizations

### 9.1 A polynomial and its derivative

Let us visualize a polynomial together with its derivative.

Take

$$
f(x)=x^3-3x^2+2
$$

Then

$$
f'(x)=3x^2-6x
$$

```
x_vals = np.linspace(-1, 4, 400)
f_vals = x_vals**3 - 3*x_vals**2 + 2
fp_vals = 3*x_vals**2 - 6*x_vals

plt.figure(figsize=(8,5))
plt.plot(x_vals, f_vals, label="f(x) = x^3 - 3x^2 + 2")
plt.plot(x_vals, fp_vals, label="f'(x) = 3x^2 - 6x")
plt.title("A polynomial and its derivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

When the derivative is positive, the function is increasing.
When the derivative is negative, the function is decreasing.
Where the derivative is zero, the graph may have a turning point or a flat point.

## 11. Numeric Check

We can compare an exact symbolic derivative with a numerical approximation.

```
def numerical_derivative(func, x, h=1e-6):
    return (func(x + h) - func(x)) / h

def poly(t):
    return 3*t**3 + 5*t**2 + 8*t + 1

x0 = 2.0
numeric = numerical_derivative(poly, x0)
exact = 9*x0**2 + 10*x0 + 8

print("Numerical derivative at x=2:", numeric)
print("Exact derivative at x=2:", exact)
```

## 14. Conclusion

In this notebook, we studied the main derivative rules used for polynomials and related algebraic expressions.

The key rules are:

$$
(c)'=0
$$

$$
(x^n)'=n x^{n-1}
$$

$$
(cf)'=cf'
$$

$$
(f+g)'=f'+g', \qquad (f-g)'=f'-g'
$$

$$
(fg)'=f'g+fg'
$$

$$
\left(\frac{f}{g}\right)'=\frac{f'g-fg'}{g^2}
$$

For polynomial functions, these rules make differentiation straightforward:
differentiate term by term using the power rule.

These tools are foundational for studying tangent lines, optimization, curve sketching, and more advanced calculus topics.
