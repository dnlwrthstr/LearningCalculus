## 1. Introduction

You already know how to differentiate:

- polynomials
- trigonometric functions
- products
- quotients

But many real functions are **nested** inside other functions, such as:

$$
f(x)=(x^5-2x+11)^7
$$

or

$$
f(x)=\tan\left(\cos(x^2)\right)^3
$$

These are called **composite functions**.

To differentiate them efficiently, we use one of the most important rules in calculus:

## The Chain Rule


## 2. Composite Functions

A composite function has one function applied **inside** another.

General form:

$$
F(x)=f(g(x))
$$

Read as:

> “f of g of x”

Where:

- $g(x)$ is the **inner function**
- $f(u)$ is the **outer function**

We also write:

$$
(f \circ g)(x)=f(g(x))
$$

where $\circ$ means composition.


## 3. Example of Composition

Consider:

$$
F(x)=(x^5-2x+11)^7
$$

We can split it into:

Inner function:

$$
g(x)=x^5-2x+11
$$

Outer function:

$$
f(u)=u^7
$$

Then:

$$
F(x)=f(g(x))
$$


## 4. The Chain Rule

If

$$
F(x)=f(g(x))
$$

then

$$
F'(x)=f'(g(x))\cdot g'(x)
$$

Meaning:

1. Differentiate the **outer function**
2. Keep the inside unchanged
3. Multiply by derivative of the **inner function**

This is the chain rule.


## 5. Worked Example 1

Find the derivative of:

$$
F(x)=(x^4+2)^{1/3}
$$

### Step 1: Identify functions

Outer:

$$
f(u)=u^{1/3}
$$

Inner:

$$
g(x)=x^4+2
$$

### Step 2: Differentiate outer

$$
f'(u)=\frac13 u^{-2/3}
$$

So:

$$
f'(g(x))=\frac13 (x^4+2)^{-2/3}
$$

### Step 3: Differentiate inner

$$
g'(x)=4x^3
$$

### Step 4: Multiply

$$
F'(x)=\frac13(x^4+2)^{-2/3}\cdot4x^3
$$

$$
F'(x)=\frac{4x^3}{3(x^4+2)^{2/3}}
$$


## 6. Worked Example 2 — Order Matters

Consider:

$$
F(x)=\cos(x^3)
$$

Outer function: cosine

Inner function:

$$
x^3
$$

Apply chain rule:

$$
F'(x)=-\sin(x^3)\cdot 3x^2
$$

$$
F'(x)=-3x^2\sin(x^3)
$$


Now compare with:

$$
G(x)=\cos^3(x)=(\cos x)^3
$$

Here:

Outer function:

$$
u^3
$$

Inner function:

$$
u=\cos x
$$

Then:

$$
G'(x)=3\cos^2(x)\cdot(-\sin x)
$$

$$
G'(x)=-3\cos^2(x)\sin x
$$

Even though the expressions look similar, the derivatives are different.

**Order matters.**


## 7. Worked Example 3 — Multiple Layers

Consider:

$$
F(x)=\cos(\sin(\cot x))
$$

There are three layers:

1. outermost: $\cos(u)$
2. middle: $\sin(v)$
3. inner: $\cot x$

Apply chain rule repeatedly:

$$
F'(x)= -\sin(\sin(\cot x))\cdot (\sin(\cot x))'
$$

Differentiate middle layer:

$$
(\sin(\cot x))'=\cos(\cot x)\cdot (\cot x)'
$$

Use:

$$
(\cot x)'=-\frac1{\sin^2 x}
$$

So:

$$
F'(x)= -\sin(\sin(\cot x))\cdot \cos(\cot x)\cdot\left(-\frac1{\sin^2 x}\right)
$$

Final result:

$$
F'(x)=\frac{\sin(\sin(\cot x))\cos(\cot x)}{\sin^2 x}
$$


```
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

## 8. Symbolic Verification with SymPy

```
x = sp.symbols('x')

exprs = [
    (x**4 + 2)**sp.Rational(1,3),
    sp.cos(x**3),
    sp.cos(x)**3,
    sp.cos(sp.sin(sp.cot(x)))
]

for expr in exprs:
    print("f(x) =", expr)
    print("f'(x) =", sp.simplify(sp.diff(expr, x)))
    print("-"*80)
```

## 9. Visualization — $\cos(x^3)$ and its Derivative

```
xv = np.linspace(-2,2,600)
y = np.cos(xv**3)
dy = -3*xv**2*np.sin(xv**3)

plt.figure(figsize=(8,5))
plt.plot(xv, y, label='cos(x^3)')
plt.plot(xv, dy, label='derivative')
plt.title("Composite Function and Derivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 10. Shortcut Strategy

When differentiating nested functions:

### Outside → Inside

Example:

$$
(3x^2+1)^5
$$

Think:

- power of 5 outside
- polynomial inside

Derivative:

$$
5(3x^2+1)^4\cdot6x
$$

This mental model works extremely well.

## 13. Conclusion

You learned:

### Composite Function

$$
f(g(x))
$$

### Chain Rule

$$
(f(g(x)))'=f'(g(x))g'(x)
$$

Core workflow:

1. Differentiate outside function
2. Keep inside unchanged
3. Multiply by derivative of inside
4. Repeat for multiple layers

The chain rule is one of the most important tools in all calculus.
