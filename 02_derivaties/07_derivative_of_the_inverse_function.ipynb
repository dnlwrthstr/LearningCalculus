# Derivative of the Inverse Function

A guided notebook on inverse functions and their derivatives.

This notebook explains:

- what an inverse function is
- how graphs of inverse functions are related
- why inverse derivatives are reciprocals
- how to use the formula
- examples with $f(x)=x^2$ and $f^{-1}(x)=\sqrt{x}$

This notebook follows the same style template as the uploaded guidelines.


## 1. Introduction

You already know how to differentiate many types of functions.

Now we study an important question:

> If a function has an inverse, how is the derivative of the inverse related to the derivative of the original function?

The key formula is:

$$
(f^{-1})'(f(a))=\frac{1}{f'(a)}
$$

Equivalently:

$$
(f^{-1})'(y)=\frac{1}{f'(x)}
\quad \text{where} \quad y=f(x)
$$


## 2. What Is an Inverse Function?

If

$$
y=f(x)
$$

then the inverse function reverses the mapping:

$$
x=f^{-1}(y)
$$

So:

$$
f^{-1}(f(x))=x
$$

and

$$
f(f^{-1}(y))=y
$$

An inverse function undoes the original function.


## 3. Example: Squaring and Square Root

Consider:

$$
f(x)=x^2
$$

If we restrict the domain to $x\geq 0$, then the inverse is:

$$
f^{-1}(x)=\sqrt{x}
$$

Examples:

$$
f\left(\frac12\right)=\frac14
$$

and

$$
f^{-1}\left(f\left(\frac12\right)\right)
=
f^{-1}\left(\frac14\right)
=
\frac12
$$

Also:

$$
f(1)=1
$$

and

$$
f^{-1}(f(1))=f^{-1}(1)=1
$$


## 4. Graphical Relationship

The graph of an inverse function is the reflection of the original graph across the line:

$$
y=x
$$

If the original graph contains the point:

$$
(a,b)
$$

then the inverse graph contains:

$$
(b,a)
$$

Example:

$$
\left(\frac12,\frac14\right)
\quad \Longleftrightarrow \quad
\left(\frac14,\frac12\right)
$$


```
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

## 5. Visualization: $f(x)=x^2$ and $f^{-1}(x)=\sqrt{x}$

```
x = np.linspace(0, 4, 500)
f = x**2

u = np.linspace(0, 4, 500)
g = np.sqrt(u)

plt.figure(figsize=(7,7))
plt.plot(x, f, label=r"$f(x)=x^2$")
plt.plot(u, g, label=r"$f^{-1}(x)=\sqrt{x}$")
plt.plot([0, 4], [0, 4], linestyle="--", label=r"$y=x$")

points_f = [(1/2, 1/4), (1, 1), (3/2, 9/4)]
points_g = [(1/4, 1/2), (1, 1), (9/4, 3/2)]

for px, py in points_f:
    plt.scatter(px, py, s=60)
    plt.annotate(f"({px:.2g}, {py:.2g})", (px, py), textcoords="offset points", xytext=(8, 8))

for px, py in points_g:
    plt.scatter(px, py, s=60)
    plt.annotate(f"({px:.2g}, {py:.2g})", (px, py), textcoords="offset points", xytext=(8, -15))

plt.title("A Function and Its Inverse")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 4)
plt.ylim(0, 4)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 6. Derivative of the Inverse Function

Suppose:

$$
y=f(x)
$$

and

$$
x=f^{-1}(y)
$$

The derivative of the inverse is:

$$
(f^{-1})'(y)=\frac{1}{f'(x)}
$$

where $y=f(x)$.

In the common point-based form:

$$
(f^{-1})'(f(a))=\frac{1}{f'(a)}
$$

Meaning:

> The slope of the inverse at the reflected point is the reciprocal of the slope of the original function.


## 7. Why the Formula Is True

Start with:

$$
f^{-1}(f(x))=x
$$

Differentiate both sides with respect to $x$.

Using the chain rule:

$$
(f^{-1})'(f(x))\cdot f'(x)=1
$$

Solving for the derivative of the inverse:

$$
(f^{-1})'(f(x))=\frac{1}{f'(x)}
$$


## 8. Worked Example: $f(x)=x^2$

Let:

$$
f(x)=x^2
$$

with domain $x>0$.

Then:

$$
f^{-1}(x)=\sqrt{x}
$$

The derivative of $f$ is:

$$
f'(x)=2x
$$

Using the inverse derivative formula:

$$
(f^{-1})'(f(a))=\frac{1}{f'(a)}
$$

Since $f(a)=a^2$:

$$
(f^{-1})'(a^2)=\frac{1}{2a}
$$

Let $y=a^2$, so $a=\sqrt{y}$:

$$
(f^{-1})'(y)=\frac{1}{2\sqrt{y}}
$$

Therefore:

$$
\frac{d}{dy}\sqrt{y}=\frac{1}{2\sqrt{y}}
$$


## 9. Numerical Examples

At $a=\frac12$:

$$
f\left(\frac12\right)=\frac14
$$

and

$$
f'\left(\frac12\right)=1
$$

Therefore:

$$
(f^{-1})'\left(\frac14\right)=1
$$

At $a=\frac32$:

$$
f\left(\frac32\right)=\frac94
$$

and

$$
f'\left(\frac32\right)=3
$$

Therefore:

$$
(f^{-1})'\left(\frac94\right)=\frac13
$$


## 10. Tangent Line Intuition

If the tangent slope of the original function is:

$$
m
$$

then the tangent slope of the inverse function at the reflected point is:

$$
\frac{1}{m}
$$

This happens because the inverse swaps horizontal and vertical changes.


```
a = 1.5
b = a**2

slope_f = 2*a
slope_inverse = 1/slope_f

x_vals = np.linspace(0, 3, 500)
f_vals = x_vals**2

u_vals = np.linspace(0, 4, 500)
inverse_vals = np.sqrt(u_vals)

x_tangent = np.linspace(a - 0.7, a + 0.7, 100)
y_tangent = b + slope_f * (x_tangent - a)

u_tangent = np.linspace(b - 1.0, b + 1.0, 100)
v_tangent = a + slope_inverse * (u_tangent - b)

plt.figure(figsize=(7,7))
plt.plot(x_vals, f_vals, label=r"$f(x)=x^2$")
plt.plot(u_vals, inverse_vals, label=r"$f^{-1}(x)=\sqrt{x}$")
plt.plot([0, 4], [0, 4], linestyle="--", label=r"$y=x$")
plt.plot(x_tangent, y_tangent, label="tangent to f")
plt.plot(u_tangent, v_tangent, label="tangent to inverse")
plt.scatter([a, b], [b, a], s=70)
plt.title("Tangents of a Function and Its Inverse")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 4)
plt.ylim(0, 4)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print("a =", a)
print("f(a) =", b)
print("slope of f at a =", slope_f)
print("slope of inverse at f(a) =", slope_inverse)
```

## 12. Another Example: Exponential and Logarithm

The inverse of:

$$
f(x)=e^x
$$

is:

$$
f^{-1}(x)=\ln x
$$

Since:

$$
f'(x)=e^x
$$

the inverse derivative formula gives:

$$
(f^{-1})'(f(a))=\frac{1}{f'(a)}
$$

So:

$$
(\ln)'(e^a)=\frac{1}{e^a}
$$

Let $y=e^a$. Then:

$$
(\ln)'(y)=\frac{1}{y}
$$

Thus:

$$
\frac{d}{dy}\ln y=\frac1y
$$


## 13. Conditions for Using the Formula

The formula

$$
(f^{-1})'(f(a))=\frac{1}{f'(a)}
$$

requires:

1. $f$ has an inverse near $a$
2. $f$ is differentiable at $a$
3. $f'(a)\neq 0$

If $f'(a)=0$, then the reciprocal is undefined.

Geometrically, a horizontal tangent in the original function corresponds to a vertical tangent in the inverse.

## 15. Solutions / Checks

1. Since $f^{-1}(4)=2$ and $f'(2)=4$:

$$
(f^{-1})'(4)=\frac14
$$

2. Since $f^{-1}(8)=2$ and $f'(x)=3x^2$:

$$
(f^{-1})'(8)=\frac{1}{f'(2)}=\frac{1}{12}
$$

3. Since $f(x)=e^x$ and $f'(x)=e^x$, if $y=e^x$, then:

$$
(\ln)'(y)=\frac{1}{y}
$$

4. If $f(x)=2x+5$, then $f'(x)=2$, so:

$$
(f^{-1})'(x)=\frac12
$$

5. The inverse swaps input and output, so horizontal and vertical changes are swapped. Therefore slope $\frac{\Delta y}{\Delta x}$ becomes $\frac{\Delta x}{\Delta y}$.

6. Because the formula uses $\frac{1}{f'(a)}$. If $f'(a)=0$, this reciprocal is undefined.


## 16. Conclusion

Main formula:

$$
(f^{-1})'(f(a))=\frac{1}{f'(a)}
$$

Equivalent form:

$$
(f^{-1})'(y)=\frac{1}{f'(x)}
\quad \text{where} \quad y=f(x)
$$

Key intuition:

- inverse functions swap inputs and outputs
- their graphs reflect across $y=x$
- slopes become reciprocals
- horizontal tangents become vertical tangents

Important examples:

$$
\frac{d}{dx}\sqrt{x}=\frac{1}{2\sqrt{x}}
$$

and

$$
\frac{d}{dx}\ln x=\frac1x
$$
