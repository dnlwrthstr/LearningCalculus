A guided notebook on functions that are **not differentiable** at some points.

This notebook explains:
- what differentiability means
- corners, cusps, vertical tangents, and discontinuities
- why $f(x)=|x|$ is not differentiable at $x=0$
- left-hand and right-hand derivatives

## 1. Introduction

A derivative measures the instantaneous slope of a function at a point.

For a function to be differentiable at a point, the graph must have one clear tangent slope there.

A function may fail to be differentiable because of:
- a corner
- a cusp
- a vertical tangent
- a discontinuity

The classic example is:

$$
f(x)=|x|
$$

This function is continuous at $x=0$, but it is **not differentiable** at $x=0$.


## 2. Definition of the Derivative

The derivative of $f$ at $x=a$ is

$$
f'(a)=\lim_{h\to 0}\frac{f(a+h)-f(a)}{h}
$$

This limit exists only if the left-hand and right-hand limits agree:

$$
\lim_{h\to 0^-}\frac{f(a+h)-f(a)}{h}
=
\lim_{h\to 0^+}\frac{f(a+h)-f(a)}{h}
$$

If these one-sided limits are different, the derivative does not exist.


## 3. The Absolute Value Function

The absolute value function is

$$
f(x)=|x|
$$

It can be written as

$$
f(x)=
\begin{cases}
x, & x\geq 0 \\
-x, & x<0
\end{cases}
$$

So:
- for $x>0$, the graph is the line $y=x$
- for $x<0$, the graph is the line $y=-x$

At $x=0$, the two lines meet in a sharp corner.


```
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

## 4. Visualization of $f(x)=|x|$

```
x = np.linspace(-5, 5, 500)
y = np.abs(x)

plt.figure(figsize=(8,5))
plt.plot(x, y, label=r"$f(x)=|x|$")
plt.scatter([0], [0], s=70, label="corner at x=0")
plt.title("Absolute Value Function")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.axhline(0, lw=1)
plt.axvline(0, lw=1)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 5. Left-Hand and Right-Hand Slopes

For $x>0$:

$$
f(x)=x
$$

so

$$
f'(x)=1
$$

For $x<0$:

$$
f(x)=-x
$$

so

$$
f'(x)=-1
$$

At $x=0$:
- the left-hand slope approaches $-1$
- the right-hand slope approaches $1$

Since these are not equal, the derivative at $x=0$ does not exist.


## 6. Direct Calculation at $x=0$

Use the derivative definition:

$$
f'(0)=\lim_{h\to 0}\frac{|0+h|-|0|}{h}
$$

Since $|0|=0$:

$$
f'(0)=\lim_{h\to 0}\frac{|h|}{h}
$$

### Right-hand limit

If $h>0$, then $|h|=h$:

$$
\frac{|h|}{h}=1
$$

So:

$$
\lim_{h\to 0^+}\frac{|h|}{h}=1
$$

### Left-hand limit

If $h<0$, then $|h|=-h$:

$$
\frac{|h|}{h}=-1
$$

So:

$$
\lim_{h\to 0^-}\frac{|h|}{h}=-1
$$

Because $-1\neq 1$, the derivative does not exist at $x=0$.


## 7. Why Corners Cause Problems

At a smooth point, zooming in on the graph makes it look more and more like one straight line.

At a corner, zooming in still shows two different directions.

For $f(x)=|x|$ at $x=0$:
- the left side has slope $-1$
- the right side has slope $1$

There is no single tangent line that represents both sides.


## 8. Visualizing Several Candidate Lines

At a smooth point, nearby secant lines approach one limiting tangent line.

At the corner of $|x|$, different directions suggest different slopes.


```
x = np.linspace(-4, 4, 500)
y = np.abs(x)

plt.figure(figsize=(8,5))
plt.plot(x, y, label=r"$f(x)=|x|$", linewidth=2)

for m in [-1, -0.5, 0, 0.5, 1]:
    plt.plot(x, m*x, alpha=0.5, label=f"slope {m}")

plt.scatter([0], [0], s=70)
plt.title("Different Lines Through the Corner")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, lw=1)
plt.axvline(0, lw=1)
plt.grid(True, alpha=0.3)
plt.ylim(-2, 4)
plt.legend()
plt.show()
```

## 9. Differentiability Implies Continuity

A useful theorem:

> If a function is differentiable at a point, then it is continuous at that point.

So differentiability is stronger than continuity.

But the reverse is not always true.

The function

$$
f(x)=|x|
$$

is continuous at $x=0$, but it is not differentiable at $x=0$.

So:

$$
\text{differentiable} \Rightarrow \text{continuous}
$$

but

$$
\text{continuous} \nRightarrow \text{differentiable}
$$


## 10. Common Reasons a Function Is Not Differentiable

### 1. Corner

Example:

$$
f(x)=|x|
$$

at $x=0$.

### 2. Cusp

Example:

$$
f(x)=|x|^{2/3}
$$

at $x=0$.

### 3. Vertical tangent

Example:

$$
f(x)=\sqrt[3]{x}
$$

at $x=0$.

### 4. Discontinuity

Example:

$$
f(x)=
\begin{cases}
1, & x<0 \\
2, & x\geq 0
\end{cases}
$$

at $x=0$.


## 11. Visual Examples

```
x = np.linspace(-3, 3, 500)
corner = np.abs(x)
cusp = np.abs(x)**(2/3)
vertical_tangent = np.cbrt(x)

plt.figure(figsize=(8,5))
plt.plot(x, corner, label=r"corner: $|x|$")
plt.plot(x, cusp, label=r"cusp: $|x|^{2/3}$")
plt.plot(x, vertical_tangent, label=r"vertical tangent: $\sqrt[3]{x}$")
plt.title("Different Types of Non-Differentiability")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, lw=1)
plt.axvline(0, lw=1)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 12. Numerical Difference Quotients

For $f(x)=|x|$ at $x=0$, the quotient is

$$
\frac{|h|}{h}
$$

For positive $h$, it equals $1$.

For negative $h$, it equals $-1$.


```
def absolute_value(t):
    return np.abs(t)

h_values = [1, 0.1, 0.01, -0.01, -0.1, -1]

for h in h_values:
    quotient = (absolute_value(h) - absolute_value(0)) / h
    print(f"h = {h:>6}, difference quotient = {quotient}")
```

## 13. Worked Example

Determine whether

$$
f(x)=|x-2|
$$

is differentiable at $x=2$.

Rewrite as:

$$
f(x)=
\begin{cases}
x-2, & x\geq 2 \\
-(x-2), & x<2
\end{cases}
$$

For $x>2$, the slope is $1$.

For $x<2$, the slope is $-1$.

The left-hand and right-hand derivatives at $x=2$ are different.

Therefore, $f(x)=|x-2|$ is not differentiable at $x=2$.

## 16. Conclusion

A function can fail to be differentiable at a point because of:

- corners
- cusps
- vertical tangents
- discontinuities

The central example was:

$$
f(x)=|x|
$$

At $x=0$:

$$
\lim_{h\to0^-}\frac{|h|}{h}=-1
$$

but

$$
\lim_{h\to0^+}\frac{|h|}{h}=1
$$

Since the one-sided derivatives are different, the derivative does not exist.

Key lesson:

$$
\text{Differentiability requires one unique tangent slope.}
$$
