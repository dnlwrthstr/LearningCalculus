Logarithmic derivatives are extremely important because they help with:

- simplifying complicated products and powers

- growth and decay modeling
- finance and continuous compounding
- advanced calculus techniques


## 2. Basic Logarithm Definitions

The **natural logarithm** is the logarithm with base $e$:

$$
\ln x = \log_e x
$$

where

$$
e \approx 2.71828
$$

A logarithm with base $a$ can be rewritten using natural logarithms:

$$
\log_a x = \frac{\ln x}{\ln a}
$$

This identity is very useful for derivatives.


## 3. Derivative of the Natural Logarithm

One of the fundamental formulas in calculus is:

$$
\frac{d}{dx}(\ln x)=\frac{1}{x}
$$

This is valid for:

$$
x>0
$$

So the graph of $\ln x$ always has slope $1/x$.


## 4. Derivative of a Logarithm with Base $a$

Using

$$
\log_a x = \frac{\ln x}{\ln a}
$$

and noting that $\ln a$ is a constant:

$$
\frac{d}{dx}(\log_a x)
=
\frac{1}{\ln a}\frac{d}{dx}(\ln x)
$$

Therefore:

$$
\frac{d}{dx}(\log_a x)=\frac{1}{x\ln a}
$$

This is valid for:

- $a>0$
- $a\neq 1$
- $x>0$


## 5. Logarithm of a Composite Function

Suppose

$$
f(x)=\ln(g(x))
$$

Then we apply the chain rule:

$$
f'(x)=\frac{g'(x)}{g(x)}
$$

This is one of the most useful derivative formulas in calculus.


 ## 6. Worked Example 1

Find:

$$
\frac{d}{dx}\left[\ln(\sin(5x))\right]
$$

### Step 1: Outer derivative

Derivative of $\ln(u)$:

$$
\frac{1}{u}
$$

So:

$$
\frac{1}{\sin(5x)}
$$

### Step 2: Multiply by derivative of inside

Derivative of $\sin(5x)$:

$$
\cos(5x)\cdot 5
$$

### Final answer

$$
\frac{d}{dx}\ln(\sin(5x))
=
\frac{5\cos(5x)}{\sin(5x)}
$$

$$
=5\cot(5x)
$$


## 7. Worked Example 2

Find:

$$
\frac{d}{dx}\left[\log_{20}(x^3)\right]
$$

Use:

$$
\frac{d}{dx}(\log_a u)=\frac{u'}{u\ln a}
$$

Here:

$$
u=x^3
\quad\Rightarrow\quad
u'=3x^2
$$

So:

$$
\frac{d}{dx}\log_{20}(x^3)
=
\frac{3x^2}{x^3\ln 20}
$$

Simplify:

$$
=\frac{3}{x\ln 20}
$$


## 8. Important Example: Derivative of $x^x$

Consider:

$$
f(x)=x^x
$$

This is unusual because the variable appears:

- in the base
- in the exponent

We use **logarithmic differentiation**.


### Step 1: Take logarithms

$$
f(x)=x^x
$$

Take $\ln$ of both sides:

$$
\ln(f(x))=\ln(x^x)
$$

Use the power rule of logarithms:

$$
\ln(f(x))=x\ln x
$$


### Step 2: Differentiate both sides

$$
\frac{d}{dx}[\ln(f(x))]=\frac{d}{dx}[x\ln x]
$$

Left side uses chain rule:

$$
\frac{f'(x)}{f(x)}
$$

Right side uses product rule:

$$
(x\ln x)'=1\cdot\ln x + x\cdot\frac1x
$$

$$
=\ln x +1
$$

So:

$$
\frac{f'(x)}{f(x)}=\ln x +1
$$


### Step 3: Solve for $f'(x)$

Multiply both sides by $f(x)$:

$$
f'(x)=f(x)(\ln x+1)
$$

Since $f(x)=x^x$:

$$
\boxed{
\frac{d}{dx}(x^x)=x^x(\ln x+1)
}
$$


## 9. General Logarithmic Differentiation

This technique is useful whenever a function contains:

- variable exponents
- many products
- roots and powers combined
- expressions like:

$$
f(x)=\frac{x^x\sqrt{x+1}}{(2x-1)^5}
$$

Taking logarithms often converts products into sums and powers into coefficients.


```
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

## 10. Symbolic Verification with SymPy

```
x = sp.symbols('x', positive=True)

exprs = [
    sp.log(x),
    sp.log(x, 20),
    sp.log(sp.sin(5*x)),
    x**x
]

for expr in exprs:
    print("f(x) =", expr)
    print("f'(x) =", sp.simplify(sp.diff(expr, x)))
    print("-"*80)
```

## 11. Visualization of $\ln x$ and Its Derivative

```
xv = np.linspace(0.1, 6, 600)

plt.figure(figsize=(8,5))
plt.plot(xv, np.log(xv), label='ln(x)')
plt.plot(xv, 1/xv, label="derivative = 1/x")
plt.title("Natural Logarithm and Its Derivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

Notice:

- $\ln x$ increases slowly
- its derivative $1/x$ decreases as $x$ grows


## 12. Visualization of $x^x$

```
xv = np.linspace(0.2, 3, 500)
y = xv**xv
dy = y*(np.log(xv)+1)

plt.figure(figsize=(8,5))
plt.plot(xv, y, label='x^x')
plt.plot(xv, dy, label="derivative")
plt.title("x^x and Its Derivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
