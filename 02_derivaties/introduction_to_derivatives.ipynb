## 1. Introduction

Derivatives help us measure **change**.

They are used to:
- find maxima and minima
- determine increasing/decreasing intervals
- model velocity in physics
- optimize profit/cost in economics

In simple terms, the derivative tells us how fast a function changes at a point.

## 2. Behaviour of a Function

Imagine a road through hilly terrain. As you move horizontally, your height changes.

If the road is modeled by $y=f(x)$:
- moving along the road changes $x$
- climbing or descending changes $f(x)$

The question is: **How steep is the road right now?**

## 3. Average Slope

Over a finite move $\Delta x$, the change in height is $\Delta f$.

$$
\Delta f = f(x+\Delta x)-f(x)
$$

Average slope over that interval:

$$
\frac{\Delta f}{\Delta x}
$$

This is called a **difference quotient**.

## 4. Why Smaller Intervals Matter

If we measure slope over a long distance, hills and valleys may cancel out.
To estimate the slope at one exact point, we shrink the interval:

$$
\Delta x \to 0
$$

That leads to the derivative.

## 5. Formal Definition of the Derivative

The derivative of $f(x)$ is:

$$
f'(x)=\lim_{\Delta x\to 0}\frac{f(x+\Delta x)-f(x)}{\Delta x}
$$

Alternative notation:

$$
\frac{df}{dx}
$$

Interpretation:
- instantaneous slope of the tangent line
- instantaneous rate of change

```
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

## 6. Visual Intuition: Secant Line Approaches Tangent

```
x = np.linspace(-1,4,400)
y = x**2

x0 = 1
for h in [1.5, 0.7, 0.25]:
    plt.figure(figsize=(8,5))
    plt.plot(x,y,label='f(x)=x^2')
    y0 = x0**2
    x1 = x0+h
    y1 = x1**2
    m = (y1-y0)/h
    line = y0 + m*(x-x0)
    plt.plot(x,line,label=f'Secant h={h:.2f}')
    plt.scatter([x0,x1],[y0,y1])
    plt.title("Secant line approaching tangent")
    plt.xlabel("x"); plt.ylabel("y")
    plt.grid(True, alpha=0.3); plt.legend()
    plt.show()
```

As $h$ gets smaller, the secant line approaches the tangent line at $x=1$.

## 7. Worked Example: Linear Function

Let:

$$f(x)=5x+8$$

Then:

$$
\Delta f = f(x+\Delta x)-f(x)
=5(x+\Delta x)+8-(5x+8)=5\Delta x
$$

So:

$$
\frac{\Delta f}{\Delta x}=5
$$

Taking the limit:

$$
f'(x)=5
$$

A linear function has constant slope everywhere.

## 8. Worked Example: Quadratic Function

Let:

$$f(x)=x^2$$

Then:

$$
\frac{(x+\Delta x)^2-x^2}{\Delta x}
=\frac{2x\Delta x+(\Delta x)^2}{\Delta x}=2x+\Delta x
$$

Now let $\Delta x\to0$:

$$
f'(x)=2x
$$

```
x=sp.symbols('x')
f=x**2
sp.diff(f,x)
```

## 9. Numerical Approximation

```
def numerical_derivative(func, x, h=1e-5):
    return (func(x+h)-func(x))/h

print("Approx derivative of x^2 at x=3:", numerical_derivative(lambda t: t**2, 3))
print("Exact derivative:", 2*3)
```

## 10. Sign of the Derivative

- If $f'(x)>0$, the function is increasing.
- If $f'(x)<0$, the function is decreasing.
- If $f'(x)=0$, the point may be a maximum, minimum, or flat point.

```
x=np.linspace(-4,4,400)
y=x**3-3*x
dy=3*x**2-3

plt.figure(figsize=(8,5))
plt.plot(x,y,label='f(x)=x^3-3x')
plt.axhline(0, lw=1)
plt.scatter([-1,1],[2,-2], s=60, label='critical points')
plt.title("Function with turning points")
plt.xlabel("x"); plt.ylabel("y")
plt.grid(True, alpha=0.3); plt.legend()
plt.show()
```

## 13. Conclusion

We introduced derivatives as the limit of average slope.

Key idea:

$$
f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}
$$

Derivatives measure instantaneous change and are foundational in calculus, science, engineering, and finance.
