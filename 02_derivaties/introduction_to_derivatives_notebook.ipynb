# Introduction to Derivatives

A compact notebook on the basic concept of derivatives.

## 1. Idea

The derivative measures **instantaneous rate of change** or **slope**.

$$
f'(x)=\lim_{h\to 0}\frac{f(x+h)-f(x)}{h}
$$


## 2. Increments

- Input change: $\Delta x$
- Output change: $\Delta f = f(x_0+\Delta x)-f(x_0)$

Average rate of change:

$$
\frac{\Delta f}{\Delta x}
$$


## 3. Example: Linear function

For

$$
f(x)=5x+8
$$

the derivative is constant:

$$
f'(x)=5
$$


```python
import sympy as sp
x,h=sp.symbols('x h')
f=5*x+8
expr=(f.subs(x,x+h)-f)/h
print("Difference quotient:", sp.simplify(expr))
print("Derivative:", sp.limit(expr,h,0))
```

## 4. Geometric meaning

The derivative is the slope of the tangent line.


```python
import numpy as np, matplotlib.pyplot as plt
def f(x): return x**2
xs=np.linspace(-3,3,400)
plt.figure(figsize=(8,5))
plt.plot(xs,f(xs),label="x^2")
x0=1
m=2*x0
tangent=m*(xs-x0)+f(x0)
plt.plot(xs,tangent,label="Tangent at x=1")
plt.scatter([x0],[f(x0)])
plt.grid(True,alpha=.3); plt.legend(); plt.show()
```

## 5. Numerical approximation

Use small $h$:

$$
f'(x) \approx \frac{f(x+h)-f(x)}{h}
$$


```python
def f(x): return x**3
x0=2
for h in [1,0.1,0.01,0.001]:
    approx=(f(x0+h)-f(x0))/h
    print(h, approx)
print("Exact derivative of x^3 at x=2 is 12")
```

## 6. Sign of derivative

- $f'(x)>0$: increasing
- $f'(x)<0$: decreasing
- $f'(x)=0$: possible max/min/flat point
