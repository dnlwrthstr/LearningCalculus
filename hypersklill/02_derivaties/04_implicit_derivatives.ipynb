## 1. Introduction

So far, you have differentiated functions written explicitly, for example:

$$
y = 6x^2 + 9x + 10
$$

Here, $y$ is directly expressed in terms of $x$.

But many important curves are given **implicitly**, meaning that $x$ and $y$ appear together in one equation. For example:

$$
x^2 + y^2 = 25
$$

This is the equation of a circle centered at the origin with radius $5$.

In such cases, solving for $y$ first may be inconvenient, messy, or impossible.

To differentiate these relations, we use **implicit differentiation**.


## 2. What Is Implicit Differentiation?

Suppose $y$ is defined implicitly by an equation involving both $x$ and $y$:

$$
F(x,y)=0
$$

Even if $y$ is not isolated, we still think of $y$ as a function of $x$.

That means when we differentiate with respect to $x$, we must remember:

- $x$ differentiates normally
- $y$ also depends on $x$
- so derivatives involving $y$ require the **chain rule**

For example:

$$
\frac{d}{dx}(y^2)=2y\frac{dy}{dx}
$$

This is the key idea of implicit differentiation.


## 3. First Example: Circle

Consider

$$
x^2+y^2=25
$$

We want to find $\dfrac{dy}{dx}$.

### Step 1: Differentiate both sides with respect to $x$

$$
\frac{d}{dx}(x^2+y^2)=\frac{d}{dx}(25)
$$

Using the sum rule:

$$
\frac{d}{dx}(x^2)+\frac{d}{dx}(y^2)=0
$$

Now compute each derivative:

$$
\frac{d}{dx}(x^2)=2x
$$

and because $y$ depends on $x$,

$$
\frac{d}{dx}(y^2)=2y\frac{dy}{dx}
$$

So we get:

$$
2x+2y\frac{dy}{dx}=0
$$

### Step 2: Solve for $\dfrac{dy}{dx}$

$$
2y\frac{dy}{dx}=-2x
$$

$$
\frac{dy}{dx}=-\frac{2x}{2y}
$$

$$
\frac{dy}{dx}=-\frac{x}{y}
$$

So the derivative is

$$
y' = \frac{dy}{dx} = -\frac{x}{y}
$$


## 4. Why the Chain Rule Appears

When differentiating $y^2$, we must treat $y$ as a function of $x$.

Think of it as:

$$
(y(x))^2
$$

The outer function is the square, and the inner function is $y(x)$.

So by the chain rule:

$$
\frac{d}{dx}(y^2)=2y\frac{dy}{dx}
$$

Similarly:

$$
\frac{d}{dx}(y^3)=3y^2\frac{dy}{dx}
$$

and

$$
\frac{d}{dx}(\sin y)=\cos y\frac{dy}{dx}
$$


## 5. Second Example

Consider:

$$
2x^3 + x^3y + y^3 = 8
$$

We want to find $\dfrac{dy}{dx}$.

### Step 1: Differentiate both sides

$$
\frac{d}{dx}(2x^3)+\frac{d}{dx}(x^3y)+\frac{d}{dx}(y^3)=\frac{d}{dx}(8)
$$

Now differentiate each term.

For the first term:

$$
\frac{d}{dx}(2x^3)=6x^2
$$

For the product $x^3y$, use the product rule:

$$
\frac{d}{dx}(x^3y)=(x^3)'y+x^3y'
$$

$$
=3x^2y+x^3y'
$$

For the third term, use the chain rule:

$$
\frac{d}{dx}(y^3)=3y^2y'
$$

So the full equation becomes:

$$
6x^2+3x^2y+x^3y'+3y^2y'=0
$$

### Step 2: Collect the $y'$ terms

$$
x^3y' + 3y^2y' = -6x^2 - 3x^2y
$$

Factor out $y'$:

$$
y'(x^3+3y^2) = -6x^2 - 3x^2y
$$

### Step 3: Solve for $y'$

$$
y' = \frac{-6x^2 - 3x^2y}{x^3+3y^2}
$$

So:

$$
\frac{dy}{dx}=\frac{-6x^2 - 3x^2y}{x^3+3y^2}
$$


## 6. General Strategy

When doing implicit differentiation:

1. Differentiate **every term** with respect to $x$
2. Treat $y$ as a function of $x$
3. Use the **chain rule** whenever you differentiate a term involving $y$
4. Use the **product rule** if $x$ and $y$ are multiplied together
5. Collect all terms involving $y'$
6. Solve for $y'$

This process works even when you cannot solve explicitly for $y$ first.


## 7. More Examples of Implicit Equations

Implicit differentiation is useful for equations such as:

$$
x^3+y^3=8
$$

or

$$
\cos(xy)=0
$$

In both cases, $y$ is mixed with $x$ in a way that makes direct solving inconvenient or difficult.


```
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

## 8. Symbolic Verification with SymPy

We can verify implicit derivatives symbolically.


```
x, y = sp.symbols('x y')
y_func = sp.Function('y')(x)

expr1 = x**2 + y_func**2 - 25
dexpr1 = sp.diff(expr1, x)

expr2 = 2*x**3 + x**3*y_func + y_func**3 - 8
dexpr2 = sp.diff(expr2, x)

print("For x^2 + y^2 = 25:")
print(dexpr1)
print("-" * 80)
print("For 2x^3 + x^3 y + y^3 = 8:")
print(dexpr2)
```

## 9. Visualization: The Circle

The equation

$$
x^2+y^2=25
$$

represents a circle of radius $5$.

Its slope at a point is

$$
\frac{dy}{dx}=-\frac{x}{y}
$$

This means the slope depends on the point $(x,y)$ on the circle.


```
x_vals = np.linspace(-5, 5, 600)
y_upper = np.sqrt(25 - x_vals**2)
y_lower = -np.sqrt(25 - x_vals**2)

plt.figure(figsize=(6,6))
plt.plot(x_vals, y_upper, label='x^2 + y^2 = 25')
plt.plot(x_vals, y_lower)
plt.axhline(0, lw=1)
plt.axvline(0, lw=1)
plt.title("Circle: x^2 + y^2 = 25")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.legend()
plt.show()
```

## 10. Tangent Slope on the Circle

Let us compute the slope at the point $(3,4)$.

Since

$$
\frac{dy}{dx}=-\frac{x}{y}
$$

we get

$$
\left.\frac{dy}{dx}\right|_{(3,4)}=-\frac{3}{4}
$$

So the tangent line slopes downward there.


```
x0, y0 = 3, 4
slope = -x0 / y0

x_line = np.linspace(0, 5, 100)
y_line = y0 + slope * (x_line - x0)

plt.figure(figsize=(6,6))
plt.plot(x_vals, y_upper, label='circle')
plt.plot(x_vals, y_lower)
plt.scatter([x0], [y0], label='point (3,4)')
plt.plot(x_line, y_line, label='tangent line')
plt.axhline(0, lw=1)
plt.axvline(0, lw=1)
plt.title("Tangent line on the circle")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.legend()
plt.show()

print("Slope at (3,4):", slope)
```

## 11. Worked Example with Trigonometric Form

Find the derivative of

$$
\cos(xy)=0
$$

Differentiate both sides with respect to $x$:

$$
\frac{d}{dx}[\cos(xy)] = 0
$$

Use the chain rule:

$$
-\sin(xy)\cdot \frac{d}{dx}(xy)=0
$$

Now differentiate $xy$ using the product rule:

$$
\frac{d}{dx}(xy)=x\frac{dy}{dx}+y
$$

So:

$$
-\sin(xy)(x y' + y)=0
$$

If $\sin(xy)\neq 0$ *), then

$$
xy' + y = 0
$$

and therefore

$$
y'=-\frac{y}{x}
$$


*) Critical observation (this is the key fix)


“If $\sin(xy) \neq 0$, then …”

But here is the important point:

From the original equation:
$\cos(xy) = 0$

$xy = \frac{\pi}{2} + k\pi$

At those points:

$\sin(xy) = \pm 1 \neq 0$

👉 So $\sin(xy)$ is never zero on the solution set.
