## 2. Starting Point: Definition of the Derivative

Recall the derivative definition:

$$
f'(x)=\lim_{\Delta x \to 0}\frac{f(x+\Delta x)-f(x)}{\Delta x}
$$

This formula can be used to derive the trigonometric derivative rules.

In practice, once the formulas are known, we usually apply them directly rather than re-deriving them each time.

## 3. Derivative of Sine

We derive the formula for the derivative of $\sin x$ directly from the definition.

Let

$$
f(x)=\sin x
$$

Then

$$
(\sin x)'=\lim_{\Delta x\to 0}\frac{\sin(x+\Delta x)-\sin x}{\Delta x}
$$

### Step 1: Expand $\sin(x+\Delta x)$

Use the angle-addition identity:

$$
\sin(a+b)=\sin a\cos b + \cos a\sin b
$$

So:

$$
\sin(x+\Delta x)=\sin x\cos(\Delta x)+\cos x\sin(\Delta x)
$$

Substitute into the derivative formula:

$$
(\sin x)'=
\lim_{\Delta x\to0}
\frac{\sin x\cos(\Delta x)+\cos x\sin(\Delta x)-\sin x}{\Delta x}
$$

### Step 2: Group terms

$$
=
\lim_{\Delta x\to0}
\frac{\sin x(\cos(\Delta x)-1)+\cos x\sin(\Delta x)}{\Delta x}
$$

Split into two limits:

$$
=
\sin x \lim_{\Delta x\to0}\frac{\cos(\Delta x)-1}{\Delta x}
+
\cos x \lim_{\Delta x\to0}\frac{\sin(\Delta x)}{\Delta x}
$$

### Step 3: Use the standard limits

Two fundamental trigonometric limits are:

$$
\lim_{h\to0}\frac{\sin h}{h}=1
$$

and

$$
\lim_{h\to0}\frac{\cos h-1}{h}=0
$$

Therefore:

$$
(\sin x)' = \sin x \cdot 0 + \cos x \cdot 1
$$

$$
(\sin x)' = \cos x
$$

### Final Result

$$
\frac{d}{dx}(\sin x)=\cos x
$$

This means: **the slope of the graph of $\sin x$ at every point $x$ equals $\cos x$.**


## 4. Derivative of Cosine

Similarly, one can derive:

$$
(\cos x)'=-\sin x
$$

So:

$$
\frac{d}{dx}(\cos x)=-\sin x
$$

## 5. Derivative of Tangent

Recall that

$$
\tan x = \frac{\sin x}{\cos x}
$$

Using the quotient rule,

$$
\left(\frac{f}{g}\right)' = \frac{f'g - fg'}{g^2}
$$

we get

$$
(\tan x)' = \left(\frac{\sin x}{\cos x}\right)'
= \frac{(\sin x)'\cos x - \sin x (\cos x)'}{\cos^2 x}
$$

Substitute the derivatives of sine and cosine:

$$
= \frac{\cos x\cdot \cos x - \sin x(-\sin x)}{\cos^2 x}
$$

$$
= \frac{\cos^2 x + \sin^2 x}{\cos^2 x}
$$

Since

$$
\sin^2 x + \cos^2 x = 1
$$

we obtain

$$
(\tan x)' = \frac{1}{\cos^2 x}
$$

This is also written as

$$
(\tan x)' = \sec^2 x
$$

## 6. Derivative of Cotangent

Similarly,

$$
\cot x = \frac{\cos x}{\sin x}
$$

and the derivative is

$$
(\cot x)' = -\frac{1}{\sin^2 x}
$$

This is also written as

$$
(\cot x)' = -\csc^2 x
$$

## 7. Summary of the Main Formulas

The most important trigonometric derivatives are:

$$
(\sin x)' = \cos x
$$

$$
(\cos x)' = -\sin x
$$

$$
(\tan x)' = \frac{1}{\cos^2 x}
$$

$$
(\cot x)' = -\frac{1}{\sin^2 x}
$$

## 8. Visual Intuition

A derivative tells us the slope of the graph.

For example:
- the slope of $\sin x$ is $\cos x$
- the slope of $\cos x$ is $-\sin x$

We can visualize each function together with its derivative.

```
x = np.linspace(-2*np.pi, 2*np.pi, 800)

plt.figure(figsize=(8,5))
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="d/dx sin(x) = cos(x)")
plt.title("sin(x) and its derivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```
x = np.linspace(-2*np.pi, 2*np.pi, 800)

plt.figure(figsize=(8,5))
plt.plot(x, np.cos(x), label="cos(x)")
plt.plot(x, -np.sin(x), label="d/dx cos(x) = -sin(x)")
plt.title("cos(x) and its derivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 9. Worked Example 1

Find the derivative of

$$
f(x)=\tan x + \cos x
$$

Using the sum rule:

$$
f'(x)=(\tan x)' + (\cos x)'
$$

$$
f'(x)=\frac{1}{\cos^2 x} - \sin x
$$

## 10. Worked Example 2

The source text's formatting is inconsistent here, so we present the example clearly as

$$
f(x)=\frac{x}{\tan x}
$$

Using the quotient rule:

$$
f'(x)=\left(\frac{x}{\tan x}\right)'
=\frac{x'\tan x - x(\tan x)'}{\tan^2 x}
$$

Substitute $x'=1$ and $(\tan x)'=\frac{1}{\cos^2 x}$:

$$
f'(x)=\frac{\tan x - x\cdot \frac{1}{\cos^2 x}}{\tan^2 x}
$$

Now use

$$
\tan^2 x = \frac{\sin^2 x}{\cos^2 x}
$$

to simplify:

$$
f'(x)=\frac{1}{\tan x} - \frac{x}{\sin^2 x}
$$

This matches the intended result from the source text.

## 11. Symbolic Checks with SymPy

We can verify the formulas symbolically.

```
x = sp.symbols('x')

expressions = [
    sp.sin(x),
    sp.cos(x),
    sp.tan(x),
    sp.cot(x),
    sp.tan(x) + sp.cos(x),
    x / sp.tan(x)
]

for expr in expressions:
    print("f(x) =", expr)
    print("f'(x) =", sp.simplify(sp.diff(expr, x)))
    print("-" * 60)
```

## 12. Numerical Approximation

We can approximate derivatives numerically using the difference quotient.

```
def numerical_derivative(func, x, h=1e-6):
    return (func(x + h) - func(x)) / h

x0 = 1.0

approx_sin = numerical_derivative(np.sin, x0)
exact_sin = np.cos(x0)

approx_cos = numerical_derivative(np.cos, x0)
exact_cos = -np.sin(x0)

print("For sin(x) at x=1")
print("Approximate derivative:", approx_sin)
print("Exact derivative      :", exact_sin)
print()

print("For cos(x) at x=1")
print("Approximate derivative:", approx_cos)
print("Exact derivative      :", exact_cos)
```

## 13. Domain Notes

Be careful with tangent and cotangent:

- $\tan x$ is undefined where $\cos x = 0$
- $\cot x$ is undefined where $\sin x = 0$

So their derivatives are only valid where the original function is defined.
