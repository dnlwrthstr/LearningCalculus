These functions are central in finance, population growth, radioactive decay, machine learning, and differential equations.


## 2. Exponential Function

A basic exponential function has the form:

$$
f(x)=a^x
$$

where:

- $a>0$
- $a\neq 1$

Here:

- $a$ is the **base**
- $x$ is the exponent


## 3. Core Derivative Formula

The derivative of $a^x$ is:

$$
\frac{d}{dx}(a^x)=a^x\ln(a)
$$

So we keep the original function and multiply by $\ln(a)$.


## 4. Special Case: Natural Exponential

When the base is $e$:

$$
f(x)=e^x
$$

Then:

$$
\frac{d}{dx}(e^x)=e^x
$$

This is why $e^x$ is so important: it is its **own derivative**.


## 5. Why This Formula Makes Sense

Using logarithms:

$$
a^x=e^{x\ln a}
$$

Differentiate:

$$
\frac{d}{dx}(a^x)=\frac{d}{dx}(e^{x\ln a})
$$

Use chain rule:

$$
=e^{x\ln a}\cdot \ln a
$$

Since $e^{x\ln a}=a^x$:

$$
(a^x)'=a^x\ln a
$$

\frac{d}{dx}(8^x)

Use the formula:

$$
(8^x)'=8^x\ln 8

## 7. Worked Example 2

Find:

$$
g(x)=3^{x^2}
$$

This is a composite function.

Outer function:

$$
3^u
$$

Inner function:

$$
u=x^2
$$

Use chain rule:

$$
\frac{d}{dx}(3^{x^2})=3^{x^2}\ln 3\cdot (x^2)'
$$

$$
=3^{x^2}\ln 3\cdot 2x
$$

Final answer:

$$
g'(x)=2x\,3^{x^2}\ln 3
$$


## 8. Worked Example 3

The source text's notation implies the intended function is:

$$
h(x)=\tan(3^x)
$$

because the derivative uses $(3^x)'=3^x\ln 3$.

Use chain rule:

$$
h'(x)=\sec^2(3^x)\cdot (3^x)'
$$

$$
=\frac{1}{\cos^2(3^x)}\cdot 3^x\ln 3
$$

So:

$$
h'(x)=\frac{3^x\ln 3}{\cos^2(3^x)}
$$


## 9. General Rule for $a^{g(x)}$

If:

$$
f(x)=a^{g(x)}
$$

then:

$$
f'(x)=a^{g(x)}\ln(a)\cdot g'(x)
$$

This is the exponential version of the chain rule.


```
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

## 10. Symbolic Verification with SymPy

```
x=sp.symbols('x', real=True)

exprs = [
    8**x,
    3**(x**2),
    sp.exp(x),
    sp.tan(3**x)
]

for expr in exprs:
    print("f(x) =", expr)
    print("f'(x) =", sp.simplify(sp.diff(expr,x)))
    print("-"*80)
```

## 11. Visualization: $2^x$ and Its Derivative

```
xv=np.linspace(-3,3,500)
y=2**xv
dy=(2**xv)*np.log(2)

plt.figure(figsize=(8,5))
plt.plot(xv,y,label='2^x')
plt.plot(xv,dy,label="derivative = 2^x ln(2)")
plt.title("Exponential Function and Its Derivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 12. Visualization: $e^x$

Since:

$$
(e^x)'=e^x
$$

the function and derivative are identical.


```
xv=np.linspace(-3,3,500)
y=np.exp(xv)

plt.figure(figsize=(8,5))
plt.plot(xv,y,label='e^x')
plt.plot(xv,y,label="derivative = e^x")
plt.title("e^x Is Its Own Derivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
