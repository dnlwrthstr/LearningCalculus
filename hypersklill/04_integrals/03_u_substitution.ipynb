## 1. Introduction

We already know how to find direct antiderivatives of common functions. But real integrands are seldom pure, unadulterated functions — they are often **composite functions**.

In this notebook we explore the relationship between the **chain rule** and the integration of composite functions, and we learn a technique — **$u$-substitution** — to reshape complicated integrands into expressions with known direct antiderivatives.

The single guiding idea: *$u$-substitution is the chain rule run backwards.*

## 2. Integration of Composite Functions

Consider the derivative of a composite function. By the **chain rule**,

$$
\frac{d}{dx}\,f(g(x)) = f'(g(x))\cdot g'(x).
$$

Reading this backwards as an integral must give

$$
\int f'(g(x))\cdot g'(x)\,dx = f(g(x)) + C.
$$

So whenever an integrand is the product of an **outer derivative evaluated at an inner function** and the **derivative of that inner function**, we can integrate it immediately.

### Example: $\displaystyle\int \tan(x)\,dx$ via the reverse chain rule

Recall that $\tan(x) = \dfrac{\sin(x)}{\cos(x)}$, so

$$
\int \tan(x)\,dx = \int \frac{1}{\cos(x)}\cdot \sin(x)\,dx.
$$

We know the logarithm rule, generalized with the chain rule:

$$
\frac{d}{dx}\ln(g(x)) = \frac{1}{g(x)}\cdot g'(x)
\;\Longrightarrow\;
\int \frac{1}{g(x)}\cdot g'(x)\,dx = \ln|g(x)| + C.
$$

Choosing $g(x) = \cos(x)$ (so $g'(x) = -\sin(x)$):

$$
\int \frac{1}{\cos(x)}\cdot(\cos(x))'\,dx = \ln|\cos(x)| + C
\;\Longrightarrow\;
-\int \frac{1}{\cos(x)}\cdot \sin(x)\,dx = -\ln|\cos(x)| + C.
$$

Therefore:

$$
\boxed{\;\int \tan(x)\,dx = -\ln|\cos(x)| + C\;}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

x, u, v = sp.symbols('x u v', real=True)
```

```python
print("integral of tan(x) =", sp.integrate(sp.tan(x), x), "+ C")
# SymPy returns -log(cos(x)); confirm by differentiating back
print("check d/dx[-ln|cos x|] =", sp.simplify(sp.diff(-sp.log(sp.cos(x)), x)))
```

## 3. The $u$-substitution Technique

In the previous example we integrated a composite function by recognizing that the derivative of the inner function appeared as the remaining factor. To find the right combination of factors systematically, we use **$u$-substitution** (a.k.a. integration by change of variables).

Consider

$$
I = \int 2x\cdot 3^{x^2 + 5}\,dx.
$$

**Define the substitution** $u = x^2 + 5$. Differentiating,

$$
\frac{du}{dx} = 2x \;\Longrightarrow\; du = 2x\,dx.
$$

Substituting into the integral,

$$
I = \int 3^{x^2 + 5}\cdot \underbrace{2x\,dx}_{du} = \int 3^{u}\,du.
$$

This has a direct antiderivative:

$$
\int 3^{u}\,du = \frac{3^{u}}{\ln(3)} + C.
$$

Substituting back $u = x^2 + 5$:

$$
\boxed{\;I = \frac{3^{\,x^2 + 5}}{\ln(3)} + C\;}
$$

> **A note on notation.** $\dfrac{du}{dx}$ is **not** a fraction. Although it looks like we multiply by $dx$ on both sides, this works only because we are reversing the chain rule; in general $du = u'(x)\,dx$ holds for single-variable differential forms.

```python
I = sp.integrate(2*x*3**(x**2 + 5), x)
print("integral of 2x * 3^(x^2+5) =", sp.simplify(I), "+ C")
```

### Revisiting $\int \tan(x)\,dx$ with $u$-substitution

$$
I = \int \tan(x)\,dx = \int \frac{\sin(x)}{\cos(x)}\,dx.
$$

Let $u = \cos(x)$. Then

$$
\frac{du}{dx} = -\sin(x) \;\Longrightarrow\; du = -\sin(x)\,dx \;\Longrightarrow\; -du = \sin(x)\,dx.
$$

So

$$
I = -\int \frac{1}{u}\,du = -\ln|u| + C = -\ln|\cos(x)| + C,
$$

matching the earlier result $\displaystyle\int \tan(x)\,dx = -\ln|\cos(x)| + C$.

## 4. $u$-substitution: the Secant Function

We can use $u$-substitution to find $\displaystyle I = \int \sec(x)\,dx$.

Recall two derivatives:

$$
\frac{d}{dx}\sec(x) = \sec(x)\tan(x), \qquad \frac{d}{dx}\tan(x) = \sec^2(x),
$$

so that

$$
\frac{d}{dx}\big(\sec(x) + \tan(x)\big) = \sec(x)\big(\tan(x) + \sec(x)\big).
$$

The trick: multiply and divide the integrand by $\big(\tan(x) + \sec(x)\big)$:

$$
\int \sec(x)\,dx = \int \sec(x)\cdot \frac{\tan(x) + \sec(x)}{\tan(x) + \sec(x)}\,dx.
$$

Now let $u = \tan(x) + \sec(x)$, so $du = \sec(x)\big(\tan(x) + \sec(x)\big)\,dx$ — exactly the numerator! Hence

$$
I = \int \frac{1}{u}\,du = \ln|u| + C = \ln|\tan(x) + \sec(x)| + C.
$$

Therefore:

$$
\boxed{\;\int \sec(x)\,dx = \ln|\tan(x) + \sec(x)| + C\;}
$$

```python
# Verify by differentiating the proposed antiderivative
F = sp.log(sp.tan(x) + sp.sec(x))
print("d/dx ln|tan(x) + sec(x)| =", sp.simplify(sp.diff(F, x)))
```

## 5. $u$-substitution: another example

Consider

$$
I = \int \frac{1}{(x-1)\cdot x}\,dx.
$$

First, some algebraic manipulation splits the fraction. Writing the numerator $1$ as $x - (x-1)$:

$$
I = \int \frac{x - (x-1)}{(x-1)\,x}\,dx
= \int \frac{x}{(x-1)\,x}\,dx - \int \frac{x-1}{(x-1)\,x}\,dx
= \int \frac{1}{x-1}\,dx - \int \frac{1}{x}\,dx.
$$

So $I = I_1 + I_2$ with $I_1 = \displaystyle\int \frac{1}{x-1}\,dx$ and $I_2 = -\displaystyle\int \frac{1}{x}\,dx = -\ln|x| + C$.

For $I_1$, substitute $u = x - 1$, so $du = dx$:

$$
I_1 = \int \frac{1}{u}\,du = \ln|u| + C = \ln|x-1| + C.
$$

Therefore:

$$
\boxed{\;I = \ln|x-1| - \ln|x| + C\;}
$$

```python
I = sp.integrate(1/((x - 1)*x), x)
print("integral of 1/((x-1)x) =", I, "+ C")
```

## 6. Two Approaches to the Substitution

Consider

$$
I = \int \frac{1}{e^x + 1}\,dx.
$$

After defining a substitution, there are **two ways** to proceed: (1) differentiate implicitly to find $du$, or (2) solve for $x$ and then differentiate to find $dx$.

### Approach 1 — differentiate implicitly to find $du$

Let $u = e^x + 1$, so $du = e^x\,dx$. The expression $e^x\,dx$ does not appear yet, so multiply and divide by $e^x$:

$$
I = \int \frac{e^x}{e^x}\cdot\frac{1}{e^x + 1}\,dx = \int \frac{e^x}{e^x(e^x + 1)}\,dx.
$$

Since $u = e^x + 1$ gives $e^x = u - 1$:

$$
I = \int \frac{1}{(u-1)\,u}\,du = \ln|u-1| - \ln|u| + C,
$$

using the result from the previous section. Reversing the substitution:

$$
\boxed{\;I = \ln(e^x) - \ln(e^x + 1) + C = x - \ln(e^x + 1) + C\;}
$$

### Approach 2 — solve for $x$ then differentiate

Let $u = e^x + 1$. Solve for $x$: $\;x = \ln(u - 1)$. Then

$$
\frac{dx}{du} = \frac{1}{u-1} \;\Longrightarrow\; dx = \frac{1}{u-1}\,du.
$$

Substituting both $x$ and $dx$:

$$
I = \int \frac{1}{u}\cdot\frac{1}{u-1}\,du = \int \frac{1}{u(u-1)}\,du = \ln|u-1| - \ln|u| + C,
$$

which is the **same integral** as before, giving the same answer. Both approaches agree.

```python
I = sp.integrate(1/(sp.exp(x) + 1), x)
print("integral of 1/(e^x + 1) =", sp.simplify(I), "+ C")
# equivalently x - ln(e^x + 1); note ln(e^x) = x
print("check: x - ln(e^x+1) differentiates to",
      sp.simplify(sp.diff(x - sp.log(sp.exp(x) + 1), x)))
```

## 7. Change of Variables: a Nested Substitution

Substitution can be applied **as many times as necessary**, as long as every change of variable is reversed at the end. Consider

$$
I = \int \frac{5}{4 + \sqrt{x + 3}}\,dx.
$$

**First substitution.** Let $u = \sqrt{x + 3}$, so $x = u^2 - 3$ and $dx = 2u\,du$:

$$
I = \int \frac{5}{4 + u}\cdot 2u\,du.
$$

**Second substitution.** Let $v = 4 + u$, so $dv = du$ and $u = v - 4$:

$$
I = \int \frac{5}{v}\cdot 2(v - 4)\,dv = 10\int \frac{v - 4}{v}\,dv = 10\left(\int dv - 4\int \frac{1}{v}\,dv\right).
$$

Both are direct antiderivatives:

$$
I = 10v - 40\ln|v| + C.
$$

**Reverse every substitution** (last to first): $v = 4 + u$, then $u = \sqrt{x+3}$:

$$
I = 10(4 + u) - 40\ln|4 + u| + C = 10\big(4 + \sqrt{x+3}\big) - 40\ln\big|4 + \sqrt{x+3}\big| + C.
$$

Absorbing the constant $40$ into $C$:

$$
\boxed{\;\int \frac{5}{4 + \sqrt{x+3}}\,dx = 10\sqrt{x+3} - 40\ln\big|4 + \sqrt{x+3}\big| + C\;}
$$

```python
x_pos = sp.symbols('x', positive=True)
I = sp.integrate(5/(4 + sp.sqrt(x_pos + 3)), x_pos)
print("SymPy result:")
print("  ", sp.simplify(I), "+ C")
# Our antiderivative differs from SymPy's only by a constant; confirm the derivative matches
F = 10*sp.sqrt(x_pos + 3) - 40*sp.log(4 + sp.sqrt(x_pos + 3))
print("d/dx of our answer =", sp.simplify(sp.diff(F, x_pos)))
print("original integrand =", sp.simplify(5/(4 + sp.sqrt(x_pos + 3))))
```

## 8. Common Pitfalls of $u$-substitution

- When choosing $u$, pick an expression that is **part of the integrand**, and make sure the resulting integral (after substituting both $u$ and $du$) is **easier** — ideally one with a direct antiderivative. If it gets harder, redefine $u$ and try again.
- The antiderivative of a composite function is **not** just the antiderivative of the outermost function. Always remember the goal: reshape the integrand into the form $f(u)\,u'\,du$.

### Illustration

$$
I = \int e^{3x^2 + 2}\cdot(2x)\,dx.
$$

**A poor choice:** $u = 2x \Rightarrow x = \tfrac{1}{2}u,\; dx = \tfrac{1}{2}\,du$, giving

$$
I = \frac{1}{2}\int e^{\frac{3}{4}u^2 + 2}\cdot u\,du,
$$

which is hardly simpler.

**A good choice:** notice $\dfrac{d}{dx}(3x^2 + 2) = 6x$, a factor already present (up to a constant). Let $u = 3x^2 + 2 \Rightarrow du = 6x\,dx$:

$$
I = \frac{1}{3}\int e^{u}\,du = \frac{1}{3}e^{u} + C = \frac{1}{3}e^{3x^2 + 2} + C,
$$

which is much easier!

```python
I = sp.integrate(sp.exp(3*x**2 + 2)*(2*x), x)
print("integral of e^(3x^2+2) * 2x =", sp.simplify(I), "+ C")
```

## 9. The $u$-substitution Recipe

1. **Choose** $u$ to be an inner expression whose derivative also appears in the integrand.
2. **Differentiate** to get $du = u'(x)\,dx$ (or solve for $x$ and find $dx$).
3. **Rewrite** the whole integral in terms of $u$ — no $x$ should remain.
4. **Integrate** using a known direct antiderivative.
5. **Substitute back** to return to $x$ (reversing every substitution, last to first).
6. (Optional) **Check** by differentiating the result.

## 11. Solutions / Checks

**1.** $u = x^2 + 1$, $du = 2x\,dx$: $\displaystyle\int u^3\,du = \frac{u^4}{4} + C = \frac{(x^2+1)^4}{4} + C$.

**2.** $u = \sin x$, $du = \cos x\,dx$: $\displaystyle\int e^u\,du = e^u + C = e^{\sin x} + C$.

**3.** $u = \sin x$, $du = \cos x\,dx$: $\displaystyle\int \frac{1}{u}\,du = \ln|\sin x| + C$.

**4.** $u = x^2 + 4$, $du = 2x\,dx$: $\displaystyle\frac{1}{2}\int \frac{1}{u}\,du = \frac{1}{2}\ln(x^2 + 4) + C$.

**5.** $u = \ln x$, $du = \frac{1}{x}\,dx$: $\displaystyle\int u^2\,du = \frac{(\ln x)^3}{3} + C$.

**6.** $u = x + 1$, $x = u - 1$, $dx = du$: $\displaystyle\int (u-1)\sqrt{u}\,du = \int (u^{3/2} - u^{1/2})\,du = \frac{2}{5}(x+1)^{5/2} - \frac{2}{3}(x+1)^{3/2} + C$.

**7.** $u = x^2$, $du = 2x\,dx$; limits $x:0\to1$ become $u:0\to1$: $\displaystyle\int_0^1 e^u\,du = e - 1 \approx 1.718$.

**8.** $\displaystyle\int \csc x\,dx = \ln|\csc x - \cot x| + C$.

Run the cell below to verify these symbolically.

## 12. Conclusion

In this topic we learned:

- Integrals whose antiderivatives are **composite functions** can be solved by applying the **chain rule in reverse**.
- The **$u$-substitution** technique simplifies complicated expressions into simpler ones, usually with direct antiderivatives.
- Substitution can be applied **as many times as necessary**, as long as every change of variable is reversed.
- There are **two approaches** once $u$ is chosen: differentiate implicitly to find $du$, or solve for $x$ and differentiate to find $dx$.

We also derived the antiderivative of the secant function:

$$
\int \sec(x)\,dx = \ln|\tan(x) + \sec(x)| + C.
$$

Key lesson:

$$
\int f'(g(x))\,g'(x)\,dx = f(g(x)) + C \quad\text{— substitution is the chain rule, reversed.}
$$
