# Trigonometric Functions

A companion notebook covering sine, cosine, tangent, cotangent, identities, graphs, and applications.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Core Definitions

For angle $\alpha$ in a right triangle:

- $\sin(\alpha)=\frac{\text{opposite}}{\text{hypotenuse}}$

- $\cos(\alpha)=\frac{\text{adjacent}}{\text{hypotenuse}}$

- $\tan(\alpha)=\frac{\text{opposite}}{\text{adjacent}}$

- $\cot(\alpha)=\frac{\text{adjacent}}{\text{opposite}}$


```python
opp, adj, hyp = 4, 3, 5
print("sin =", opp/hyp)
print("cos =", adj/hyp)
print("tan =", opp/adj)
print("cot =", adj/opp)
```

## Common Angles

```python
angles = [0, np.pi/6, np.pi/4, np.pi/3, np.pi/2]
rows=[]
for a in angles:
    rows.append([a, np.sin(a), np.cos(a)])
pd.DataFrame(rows, columns=["radians","sin","cos"])
```

## Pythagorean Identity

$\sin^2(x)+\cos^2(x)=1$

```python
xs=np.linspace(-3,3,8)
for x in xs:
    print(round(np.sin(x)**2 + np.cos(x)**2,10))
```

## Graph of Sine and Cosine

```python
x=np.linspace(-2*np.pi,2*np.pi,800)
plt.figure(figsize=(9,5))
plt.plot(x,np.sin(x),label="sin(x)")
plt.plot(x,np.cos(x),label="cos(x)")
plt.axhline(0); plt.axvline(0)
plt.legend()
plt.grid(True,alpha=0.3)
plt.show()
```

## Cable Car Example

Height = length $\cdot \sin(\theta)$

```python
length=1200
theta=np.deg2rad(30)
print("Height =", length*np.sin(theta))
```

## Velocity Components

$V_x=V\cos(\alpha),\;V_y=V\sin(\alpha)$

```python
V=20
a=np.deg2rad(35)
print("Vx =", V*np.cos(a))
print("Vy =", V*np.sin(a))
```
