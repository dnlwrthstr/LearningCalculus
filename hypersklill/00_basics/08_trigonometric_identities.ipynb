# More Trigonometric Identities

This notebook is a companion to the trigonometric functions notebook.

It focuses on important **trigonometric identities**, including:

- periodicity
- the Pythagorean identity
- complementary-angle identities
- identities involving tangent and cotangent
- odd/even symmetry identities
- angle sum and difference formulas
- double-angle formulas
- a few product / quotient relations
- numerical verification in Python

These identities are useful for simplifying expressions, solving equations, and finding unknown trigonometric values.


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## 1. Periodicity of sine and cosine

Sine and cosine are periodic functions with period $2\pi$:

$$
\sin(x+2\pi)=\sin(x)
$$

$$
\cos(x+2\pi)=\cos(x)
$$

This means their values repeat every full turn around the unit circle.


```python
xs = [0.2, 1.1, 2.7, -0.9]

for x in xs:
    print(f"x = {x}")
    print("sin(x)     =", np.sin(x))
    print("sin(x+2π)  =", np.sin(x + 2*np.pi))
    print("cos(x)     =", np.cos(x))
    print("cos(x+2π)  =", np.cos(x + 2*np.pi))
    print()
```

## 2. The fundamental Pythagorean identity

The most important trigonometric identity is:

$$
\sin^2(x)+\cos^2(x)=1
$$

This follows from the unit circle equation

$$
x^2+y^2=1
$$

when we identify

$$
x=\cos(\theta), \qquad y=\sin(\theta).
$$


```python
test_values = np.linspace(-3, 3, 8)

for x in test_values:
    lhs = np.sin(x)**2 + np.cos(x)**2
    print(f"x = {x: .3f},  sin^2(x)+cos^2(x) = {lhs:.10f}")
```

## 3. Recovering sine from cosine, and cosine from sine

From

$$
\sin^2(x)+\cos^2(x)=1
$$

we get:

$$
\sin^2(x)=1-\cos^2(x)
$$

and

$$
\cos^2(x)=1-\sin^2(x)
$$

So:

$$
\sin(x)=\pm\sqrt{1-\cos^2(x)}
$$

$$
\cos(x)=\pm\sqrt{1-\sin^2(x)}
$$

The sign depends on the quadrant of the angle.


```python
# Example: if cos(x)=3/5 and x is in Quadrant I, then sin(x)>0
cos_x = 3/5
sin_x = np.sqrt(1 - cos_x**2)

print("Given cos(x)=3/5 in Quadrant I:")
print("sin(x) =", sin_x)

# Example: if cos(x)=3/5 and x is in Quadrant IV, then sin(x)<0
print("If x were in Quadrant IV instead:")
print("sin(x) =", -sin_x)
```

## 4. Complementary-angle identities

These are the identities you pasted and they are extremely useful:

$$
\sin\left(\frac{\pi}{2}-\alpha\right)=\cos(\alpha)
$$

$$
\cos\left(\frac{\pi}{2}-\alpha\right)=\sin(\alpha)
$$

$$
\tan\left(\frac{\pi}{2}-\alpha\right)=\cot(\alpha)
$$

$$
\cot\left(\frac{\pi}{2}-\alpha\right)=\tan(\alpha)
$$


```python
alpha = np.pi / 6

print("alpha =", alpha)
print()
print("sin(pi/2 - alpha) =", np.sin(np.pi/2 - alpha))
print("cos(alpha)        =", np.cos(alpha))
print()
print("cos(pi/2 - alpha) =", np.cos(np.pi/2 - alpha))
print("sin(alpha)        =", np.sin(alpha))
print()
print("tan(pi/2 - alpha) =", np.tan(np.pi/2 - alpha))
print("cot(alpha)        =", 1 / np.tan(alpha))
print()
print("cot(pi/2 - alpha) =", 1 / np.tan(np.pi/2 - alpha))
print("tan(alpha)        =", np.tan(alpha))
```

## 5. Pythagorean identities for tangent and cotangent

Starting from

$$
\sin^2(x)+\cos^2(x)=1
$$

we can derive:

$$
1+\tan^2(x)=\frac{1}{\cos^2(x)}
$$

and

$$
1+\cot^2(x)=\frac{1}{\sin^2(x)}
$$

These are very useful when tangent or cotangent is known.


```python
x = 0.7

lhs1 = 1 + np.tan(x)**2
rhs1 = 1 / np.cos(x)**2

lhs2 = 1 + (1/np.tan(x))**2
rhs2 = 1 / np.sin(x)**2

print("1 + tan^2(x) =", lhs1)
print("1 / cos^2(x) =", rhs1)
print()
print("1 + cot^2(x) =", lhs2)
print("1 / sin^2(x) =", rhs2)
```

## 6. Recovering cosine and sine from tangent

Suppose:

$$
\tan(x)=2
$$

and

$$
0 \le x \le \frac{\pi}{2}
$$

Then both sine and cosine are nonnegative.

Using

$$
1+\tan^2(x)=\frac{1}{\cos^2(x)}
$$

we get:

$$
\cos(x)=\frac{1}{\sqrt{1+2^2}}=\frac{1}{\sqrt{5}}
$$

Then:

$$
\sin(x)=\tan(x)\cos(x)=2\cdot\frac{1}{\sqrt{5}}=\frac{2}{\sqrt{5}}
$$


```python
tan_x = 2
cos_x = 1 / np.sqrt(1 + tan_x**2)
sin_x = tan_x * cos_x

print("Given tan(x)=2 in Quadrant I:")
print("cos(x) =", cos_x)
print("sin(x) =", sin_x)
print("check tan(x) =", sin_x / cos_x)
```

## 7. Odd/even symmetry identities

Sine, cosine, tangent, and cotangent have symmetry properties:

$$
\sin(-x)=-\sin(x)
$$

$$
\cos(-x)=\cos(x)
$$

$$
\tan(-x)=-\tan(x)
$$

$$
\cot(-x)=-\cot(x)
$$

So:

- sine is **odd**
- cosine is **even**
- tangent is **odd**
- cotangent is **odd**


```python
x = 0.9

print("sin(-x) =", np.sin(-x), "   -sin(x) =", -np.sin(x))
print("cos(-x) =", np.cos(-x), "    cos(x) =", np.cos(x))
print("tan(-x) =", np.tan(-x), "   -tan(x) =", -np.tan(x))
print("cot(-x) =", 1/np.tan(-x), "   -cot(x) =", -(1/np.tan(x)))
```

## 8. Sum and difference formulas

These are standard identities that appear everywhere in trigonometry:

### Sine
$$
\sin(a+b)=\sin a\cos b + \cos a\sin b
$$

$$
\sin(a-b)=\sin a\cos b - \cos a\sin b
$$

### Cosine
$$
\cos(a+b)=\cos a\cos b - \sin a\sin b
$$

$$
\cos(a-b)=\cos a\cos b + \sin a\sin b
$$


```python
a = 0.8
b = 0.3

print("sin(a+b) direct      =", np.sin(a+b))
print("sin(a+b) by identity =", np.sin(a)*np.cos(b) + np.cos(a)*np.sin(b))
print()

print("sin(a-b) direct      =", np.sin(a-b))
print("sin(a-b) by identity =", np.sin(a)*np.cos(b) - np.cos(a)*np.sin(b))
print()

print("cos(a+b) direct      =", np.cos(a+b))
print("cos(a+b) by identity =", np.cos(a)*np.cos(b) - np.sin(a)*np.sin(b))
print()

print("cos(a-b) direct      =", np.cos(a-b))
print("cos(a-b) by identity =", np.cos(a)*np.cos(b) + np.sin(a)*np.sin(b))
```

## 9. Tangent sum and difference formulas

For tangent:

$$
\tan(a+b)=\frac{\tan a + \tan b}{1-\tan a\tan b}
$$

$$
\tan(a-b)=\frac{\tan a - \tan b}{1+\tan a\tan b}
$$

These are valid where the denominators are not zero.


```python
a = 0.5
b = 0.2

lhs_plus = np.tan(a+b)
rhs_plus = (np.tan(a) + np.tan(b)) / (1 - np.tan(a)*np.tan(b))

lhs_minus = np.tan(a-b)
rhs_minus = (np.tan(a) - np.tan(b)) / (1 + np.tan(a)*np.tan(b))

print("tan(a+b) direct      =", lhs_plus)
print("tan(a+b) by identity =", rhs_plus)
print()
print("tan(a-b) direct      =", lhs_minus)
print("tan(a-b) by identity =", rhs_minus)
```

## 10. Double-angle formulas

These follow from the sum formulas by setting $a=b=x$.

### Sine
$$
\sin(2x)=2\sin(x)\cos(x)
$$

### Cosine
$$
\cos(2x)=\cos^2(x)-\sin^2(x)
$$

Also:

$$
\cos(2x)=2\cos^2(x)-1
$$

$$
\cos(2x)=1-2\sin^2(x)
$$

### Tangent
$$
\tan(2x)=\frac{2\tan(x)}{1-\tan^2(x)}
$$


```python
x = 0.4

print("sin(2x) direct      =", np.sin(2*x))
print("sin(2x) by identity =", 2*np.sin(x)*np.cos(x))
print()

print("cos(2x) direct      =", np.cos(2*x))
print("cos^2-sin^2         =", np.cos(x)**2 - np.sin(x)**2)
print("2cos^2-1            =", 2*np.cos(x)**2 - 1)
print("1-2sin^2            =", 1 - 2*np.sin(x)**2)
print()

print("tan(2x) direct      =", np.tan(2*x))
print("tan(2x) by identity =", (2*np.tan(x)) / (1 - np.tan(x)**2))
```

## 11. Half-angle style rearrangements

From the cosine double-angle identities, we can rearrange:

$$
\sin^2(x)=\frac{1-\cos(2x)}{2}
$$

$$
\cos^2(x)=\frac{1+\cos(2x)}{2}
$$

These are often helpful in integration and simplification.


```python
x = 0.9

print("sin^2(x) direct      =", np.sin(x)**2)
print("(1-cos(2x))/2        =", (1 - np.cos(2*x))/2)
print()

print("cos^2(x) direct      =", np.cos(x)**2)
print("(1+cos(2x))/2        =", (1 + np.cos(2*x))/2)
```

## 12. Product/quotient relations

Some very basic but constantly used relations are:

$$
\tan(x)=\frac{\sin(x)}{\cos(x)}
$$

$$
\cot(x)=\frac{\cos(x)}{\sin(x)}
$$

and therefore

$$
\tan(x)\cot(x)=1
$$

whenever both sides are defined.


```python
x = 1.1

tan_x = np.tan(x)
cot_x = 1 / np.tan(x)

print("tan(x) =", tan_x)
print("sin(x)/cos(x) =", np.sin(x)/np.cos(x))
print()

print("cot(x) =", cot_x)
print("cos(x)/sin(x) =", np.cos(x)/np.sin(x))
print()

print("tan(x)*cot(x) =", tan_x * cot_x)
```

## 13. Plotting periodicity

The next plot shows how sine repeats every $2\pi$.


```python
x = np.linspace(-4*np.pi, 4*np.pi, 1200)
y = np.sin(x)

plt.figure(figsize=(10,5))
plt.plot(x, y)
plt.axhline(0)
plt.axvline(0)
plt.title("Periodicity of sin(x)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True, alpha=0.3)
plt.show()
```

## 14. Using identities to solve a quick problem

Suppose:

$$
\sin(x)=\frac{4}{5}
$$

and $x$ is in Quadrant I.

Then:

$$
\cos(x)=\sqrt{1-\sin^2(x)}=\sqrt{1-\frac{16}{25}}=\frac{3}{5}
$$

and

$$
\tan(x)=\frac{\sin(x)}{\cos(x)}=\frac{4/5}{3/5}=\frac{4}{3}
$$


```python
sin_x = 4/5
cos_x = np.sqrt(1 - sin_x**2)
tan_x = sin_x / cos_x

print("Given sin(x)=4/5 in Quadrant I:")
print("cos(x) =", cos_x)
print("tan(x) =", tan_x)
```

## 15. A compact identity sheet

Here is a compact summary.

### Periodicity
$$
\sin(x+2\pi)=\sin(x), \qquad \cos(x+2\pi)=\cos(x)
$$

### Pythagorean
$$
\sin^2(x)+\cos^2(x)=1
$$

$$
1+\tan^2(x)=\frac1{\cos^2(x)}
$$

$$
1+\cot^2(x)=\frac1{\sin^2(x)}
$$

### Complementary angles
$$
\sin\left(\frac\pi2-x\right)=\cos(x)
$$

$$
\cos\left(\frac\pi2-x\right)=\sin(x)
$$

$$
\tan\left(\frac\pi2-x\right)=\cot(x)
$$

### Odd/even
$$
\sin(-x)=-\sin(x), \quad \cos(-x)=\cos(x), \quad \tan(-x)=-\tan(x)
$$

### Sum and difference
$$
\sin(a\pm b)=\sin a\cos b \pm \cos a\sin b
$$

$$
\cos(a\pm b)=\cos a\cos b \mp \sin a\sin b
$$

### Double-angle
$$
\sin(2x)=2\sin x\cos x
$$

$$
\cos(2x)=\cos^2x-\sin^2x=2\cos^2x-1=1-2\sin^2x
$$
