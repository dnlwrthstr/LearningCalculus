## 1. One-sided limits

Sometimes the value a function approaches depends on whether we come from the **left** or from the **right**.

We write:

- left-hand limit:
  $$
  \lim_{x\to a^-} f(x)
  $$

- right-hand limit:
  $$
  \lim_{x\to a^+} f(x)
  $$

A usual two-sided limit

$$
\lim_{x\to a} f(x)
$$

exists **if and only if** both one-sided limits exist and are equal.


## 2. First example

Consider the function

$$
f(x)=
\begin{cases}
-1, & x<0 \\
1, & x\ge 0
\end{cases}
$$

Then:

- approaching 0 from the left, the function stays at $-1$
- approaching 0 from the right, the function stays at $1$

So:

$$
\lim_{x\to 0^-} f(x)=-1
$$

and

$$
\lim_{x\to 0^+} f(x)=1
$$

Since these are different, the two-sided limit does **not** exist.


```python
import numpy as np
import matplotlib.pyplot as plt

def f_piecewise(x):
    return np.where(x < 0, -1, 1)

x = np.linspace(-2, 2, 600)
y = f_piecewise(x)

plt.figure(figsize=(8,5))
plt.plot(x, y)
plt.axhline(0)
plt.axvline(0)
plt.ylim(-1.5, 1.5)
plt.title("Piecewise function with different one-sided limits")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Interpreting one-sided limits

For the function above:

- if $x$ is slightly smaller than 0, then $f(x)=-1$
- if $x$ is slightly larger than 0, then $f(x)=1$

So near the same point, the function behaves differently from the two sides.


```python
xs_left = [-0.5, -0.1, -0.01, -0.001]
xs_right = [0.5, 0.1, 0.01, 0.001]

print("Approaching 0 from the left:")
for x in xs_left:
    print(f"x = {x}, f(x) = {f_piecewise(np.array([x]))[0]}")

print()
print("Approaching 0 from the right:")
for x in xs_right:
    print(f"x = {x}, f(x) = {f_piecewise(np.array([x]))[0]}")
```

## 4. Formal definitions of one-sided limits

A number $A_1$ is the **left-hand limit** of $f(x)$ at $x_0$ if:

> for every $\varepsilon > 0$, there exists $\delta > 0$ such that for all
 $$
 x \in (x_0-\delta, x_0),
 $$
 we have
 $$
 |f(x)-A_1|<\varepsilon.
 $$

A number $A_2$ is the **right-hand limit** of $f(x)$ at $x_0$ if:

> for every $\varepsilon > 0$, there exists $\delta > 0$ such that for all
 $$
 x \in (x_0, x_0+\delta),
 $$
 we have
 $$
 |f(x)-A_2|<\varepsilon.
 $$


## 5. Criterion for existence of a two-sided limit

The limit

$$
\lim_{x\to a} f(x)
$$

exists if and only if:

1. the left-hand limit exists,
2. the right-hand limit exists,
3. and they are equal.


## 6. Example where the two-sided limit exists

Consider

$$
g(x)=x^2.
$$

Then:

$$
\lim_{x\to 2^-} g(x)=4
$$

and

$$
\lim_{x\to 2^+} g(x)=4.
$$

So:

$$
\lim_{x\to 2} g(x)=4.
$$


```python
def g(x):
    return x**2

x = np.linspace(0, 4, 400)
plt.figure(figsize=(8,5))
plt.plot(x, g(x))
plt.axhline(0)
plt.axvline(0)
plt.title(r"Function $g(x)=x^2$")
plt.xlabel("x")
plt.ylabel("g(x)")
plt.grid(True, alpha=0.3)
plt.show()

for val in [1.9, 1.99, 2.01, 2.1]:
    print(f"x = {val}, g(x) = {g(val)}")
```

## 7. Non-existence of a limit

A limit may fail to exist for different reasons.

Two common reasons are:

- the left-hand and right-hand limits are different
- the function oscillates too much near the point


## 8. Heine's criterion

A number $A$ is the limit of $f(x)$ at $x_0$ if:

> for every sequence $(x_n)$ such that
 $$
 x_n \to x_0
 $$
 and
 $$
 x_n \ne x_0,
 $$
 the sequence
 $$
 f(x_n)
 $$
 tends to $A$.

This is called the **Heine definition** of the limit.


## 9. Oscillating example: $\sin(1/x)$

Consider the function

$$
f(x)=\sin\left(\frac{1}{x}\right).
$$

We want to study:

$$
\lim_{x\to 0}\sin\left(\frac{1}{x}\right).
$$

This limit does **not** exist.


```python
def s(x):
    return np.sin(1/x)

x_left = np.linspace(-0.2, -0.01, 3000)
x_right = np.linspace(0.01, 0.2, 3000)

plt.figure(figsize=(9,5))
plt.plot(x_left, s(x_left))
plt.plot(x_right, s(x_right))
plt.axhline(0)
plt.axvline(0)
plt.title(r"Oscillation of $\sin(1/x)$ near $x=0$")
plt.xlabel("x")
plt.ylabel(r"$\sin(1/x)$")
plt.grid(True, alpha=0.3)
plt.show()
```

## 10. Why $\sin(1/x)$ has no limit at 0

Approach 0 using the sequence

$$
x_n = \frac{1}{2\pi n}, \qquad n=1,2,3,\dots
$$

Then:

$$
\sin\left(\frac{1}{x_n}\right)=\sin(2\pi n)=0.
$$

Now approach 0 using

$$
y_n = \frac{1}{\pi/2 + 2\pi n}.
$$

Then:

$$
\sin\left(\frac{1}{y_n}\right)=\sin\left(\frac{\pi}{2}+2\pi n\right)=1.
$$

So two different sequences approaching the same point give two different function-value limits.

Therefore the limit does not exist.

```python
n = np.arange(1, 8)

x_n = 1 / (2*np.pi*n)
y_n = 1 / (np.pi/2 + 2*np.pi*n)

print("Sequence x_n = 1/(2πn):")
for val in x_n:
    print(val, np.sin(1/val))

print()
print("Sequence y_n = 1/(π/2 + 2πn):")
for val in y_n:
    print(val, np.sin(1/val))
```

## 11. Infinite limits

Sometimes the function does not approach a finite number at all, but instead grows without bound.

Example:

$$
\frac{1}{x}
$$

near $x=0$.

Then:

$$
\lim_{x\to 0^+}\frac{1}{x}=+\infty
$$

and

$$
\lim_{x\to 0^-}\frac{1}{x}=-\infty.
$$

In this situation, the function becomes arbitrarily large in absolute value near the point.


```python
def inv(x):
    return 1/x

x_left = np.linspace(-1, -0.02, 500)
x_right = np.linspace(0.02, 1, 500)

plt.figure(figsize=(8,5))
plt.plot(x_left, inv(x_left))
plt.plot(x_right, inv(x_right))
plt.axhline(0)
plt.axvline(0)
plt.ylim(-20, 20)
plt.title(r"Function $1/x$ near 0")
plt.xlabel("x")
plt.ylabel(r"$1/x$")
plt.grid(True, alpha=0.3)
plt.show()
```

## 12. Definitions of infinite limits

### Limit $+\infty$

We write

$$
\lim_{x\to a} f(x)=+\infty
$$

if:

> for every $N>0$, there exists $\delta>0$ such that
 $$
 0<|x-a|<\delta \implies f(x)>N.
 $$

### Limit $-\infty$

We write

$$
\lim_{x\to a} f(x)=-\infty
$$

if:

> for every $N<0$, there exists $\delta>0$ such that
 $$
 0<|x-a|<\delta \implies f(x)<N.
 $$

### Limit $\infty$ in absolute value

We write

$$
\lim_{x\to a} f(x)=\infty
$$

informally when:

> for every $N>0$, there exists $\delta>0$ such that
 $$
 0<|x-a|<\delta \implies |f(x)|>N.
 $$


## 13. One-sided infinite limits for $1/x$

For

$$
f(x)=\frac{1}{x},
$$

we have:

- from the right:
  $$
  \lim_{x\to 0^+}\frac{1}{x}=+\infty
  $$

- from the left:
  $$
  \lim_{x\to 0^-}\frac{1}{x}=-\infty
  $$

So the two sides behave differently.


```python
xs_right = [0.5, 0.1, 0.01, 0.001]
xs_left = [-0.5, -0.1, -0.01, -0.001]

print("Approaching 0 from the right:")
for x in xs_right:
    print(f"x = {x}, 1/x = {1/x}")

print()
print("Approaching 0 from the left:")
for x in xs_left:
    print(f"x = {x}, 1/x = {1/x}")
```

## 14. A function with matching infinite behavior in absolute value

Even when the left and right sides have opposite signs, the absolute value can still blow up.

For example, for

$$
\frac{1}{x},
$$

we have:

$$
\left|\frac{1}{x}\right| \to \infty
$$

as $x \to 0$.


```python
x_vals = [0.1, 0.01, -0.1, -0.01]
for x in x_vals:
    print(f"x = {x}, |1/x| = {abs(1/x)}")
```

## 15. Another example: removable vs non-removable behavior

Compare these two functions near $x=1$:

1. removable hole:
$$
\frac{x^2+x-2}{x-1}
$$

2. vertical blow-up:
$$
\frac{1}{x-1}
$$

The first has a finite limit.
The second has one-sided infinite limits.


```python
x1 = np.linspace(-1, 3, 1000)

mask = np.abs(x1 - 1) > 1e-3

y_removable = (x1[mask]**2 + x1[mask] - 2) / (x1[mask] - 1)
y_blowup = 1 / (x1[mask] - 1)

plt.figure(figsize=(8,5))
plt.plot(x1[mask], y_removable, label=r'$(x^2+x-2)/(x-1)$')
plt.scatter([1], [3], s=60, facecolors='none', edgecolors='red')
plt.axhline(0)
plt.axvline(0)
plt.ylim(-10, 10)
plt.title("Removable discontinuity")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

plt.figure(figsize=(8,5))
plt.plot(x1[mask], y_blowup, label=r'$1/(x-1)$')
plt.axhline(0)
plt.axvline(0)
plt.ylim(-10, 10)
plt.title("Infinite one-sided limits")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 16. Summary logic

A two-sided limit exists exactly when:

- the left-hand limit exists,
- the right-hand limit exists,
- and they are equal.

A two-sided limit does **not** exist when:

- the one-sided limits are different, or
- the function oscillates without settling, or
- the function blows up in incompatible ways from the two sides.
