# Sets: Notation, Concepts, and Python Examples

This notebook introduces the mathematical idea of a **set**, explains the most common **notations**, and shows how to work with sets in **Python**.

A set is a collection of **distinct** objects, called **elements**.

Examples:

- $A = \{1,2,3\}$
- $B = \{\text{apple}, \text{banana}, \text{pear}\}$

Important idea:

- In a set, **order does not matter**.
- In a set, **duplicates are ignored**.

So:

$$
\{1,2,3\} = \{3,2,1\}
$$

and

$$
\{1,1,2,2,3\} = \{1,2,3\}.
$$


## 1. Basic notation

Let $A = \{1,2,3\}$.

### Membership

- $1 \in A$ means: **1 is an element of $A$**.
- $5 \notin A$ means: **5 is not an element of $A$**.

### Equality of sets

Two sets are equal if they contain exactly the same elements.

$$
\{1,2,3\} = \{3,2,1\}
$$

### Cardinality

The number of elements in a set is called its **cardinality**.

If $A = \{1,2,3\}$, then

$$
|A| = 3.
$$

### Empty set

The set with no elements is called the **empty set**:

$$
\varnothing \quad \text{or} \quad \{\}
$$

### Set-builder notation

Sometimes a set is described by a property instead of listing elements.

Example:

$$
E = \{x \in \mathbb{N} : x \text{ is even and } x < 10\}
$$

This means:

> $E$ is the set of natural numbers that are even and less than 10.

So:

$$
E = \{2,4,6,8\}.
$$


```python
# Basic Python set examples
A = {1, 2, 3}
print('A =', A)
print('Is 1 in A?', 1 in A)
print('Is 5 in A?', 5 in A)
print('Cardinality of A:', len(A))

empty_set = set()   # not {} because {} creates a dictionary in Python
print('Empty set:', empty_set)
print('Cardinality of empty set:', len(empty_set))
```

- $\mathbb{Z}$: integers, for example $\{\dots,-2,-1,0,1,2,\dots\}$
- $\mathbb{Q}$: rational numbers

## 3. Subsets and supersets

Let

$$
A = \{1,2\}, \qquad B = \{1,2,3,4\}.
$$

### Subset

$$
A \subseteq B
$$

means: every element of $A$ is also an element of $B$.

### Proper subset

$$
A \subset B
$$

means: $A$ is a subset of $B$, and $A \neq B$.

### Superset

$$
B \supseteq A
$$

means: $B$ contains all elements of $A$.

### Important fact

Every set is a subset of itself:

$$
A \subseteq A.
$$

Also, the empty set is a subset of every set:

$$
\varnothing \subseteq A.
$$


```python
A = {1, 2}
B = {1, 2, 3, 4}

print('A =', A)
print('B =', B)
print('A <= B  (subset):', A <= B)
print('A <  B  (proper subset):', A < B)
print('B >= A  (superset):', B >= A)
print('Empty set <= A:', set() <= A)
```

## 4. Union, intersection, difference

Let

$$
A = \{1,2,3\}, \qquad B = \{3,4,5\}.
$$

### Union

The **union** contains all elements that are in $A$ or in $B$ (or in both):

$$
A \cup B = \{1,2,3,4,5\}
$$

### Intersection

The **intersection** contains the elements common to both sets:

$$
A \cap B = \{3\}
$$

### Difference

The **difference** $A \setminus B$ contains elements that are in $A$ but not in $B$:

$$
A \setminus B = \{1,2\}
$$

Similarly,

$$
B \setminus A = \{4,5\}.
$$


```python
A = {1, 2, 3}
B = {3, 4, 5}

print('A union B       =', A | B)
print('A intersection B =', A & B)
print('A - B            =', A - B)
print('B - A            =', B - A)
```

## 5. Symmetric difference

The **symmetric difference** contains elements that are in exactly one of the two sets.

$$
A \triangle B = (A \setminus B) \cup (B \setminus A)
$$

For

$$
A = \{1,2,3\}, \qquad B = \{3,4,5\},
$$

we get

$$
A \triangle B = \{1,2,4,5\}.
$$


```python
A = {1, 2, 3}
B = {3, 4, 5}

print('Symmetric difference:', A ^ B)
```

## 6. Disjoint sets

Two sets are **disjoint** if they have no common elements.

$$
A \cap B = \varnothing
$$

Example:

$$
\{1,2\} \cap \{3,4\} = \varnothing
$$


```python
A = {1, 2}
B = {3, 4}

print('A and B are disjoint:', A.isdisjoint(B))
```

## 7. Power set

The **power set** of a set $A$ is the set of all subsets of $A$.

If

$$
A = \{1,2\},
$$

then its power set is

$$
\mathcal{P}(A) = \{\varnothing, \{1\}, \{2\}, \{1,2\}\}.
$$

If a set has $n$ elements, then its power set has $2^n$ elements.


```python
from itertools import chain, combinations

def power_set(iterable):
    items = list(iterable)
    return list(chain.from_iterable(combinations(items, r) for r in range(len(items) + 1)))

A = {1, 2}
ps = power_set(A)
print('A =', A)
print('Power set of A:')
for subset in ps:
    print(set(subset))
```

## 8. Cartesian product

For two sets $A$ and $B$, the **Cartesian product** is the set of ordered pairs:

$$
A \times B = \{(a,b) : a \in A,\ b \in B\}.
$$

Example:

If

$$
A = \{1,2\}, \qquad B = \{x,y\},
$$

then

$$
A \times B = \{(1,x),(1,y),(2,x),(2,y)\}.
$$

Note that **ordered pairs are not sets**. Order matters here.

$$
(1,x) \neq (x,1).
$$


```python
A = {1, 2, 3}
B = {'x', 'y'}

cartesian = {(a, b) for a in A for b in B}
print('A x B =', cartesian)
```

## 9. Finite and infinite sets

- A **finite** set has a limited number of elements.
- An **infinite** set has infinitely many elements.

Examples:

- Finite: $\{1,2,3\}$
- Infinite: $\mathbb{N}$

In Python, we usually work with finite sets because a Python set must be stored in memory.


## 10. Common logical language in set theory

When reading set notation, these words appear often:

- **for all**: $\forall$
- **there exists**: $\exists$
- **such that**: often written as $:$ or $|$

Example:

$$
A = \{x \in \mathbb{R} : x^2 = 1\}
$$

This reads:

> $A$ is the set of all real numbers $x$ such that $x^2 = 1$.

So:

$$
A = \{-1, 1\}.
$$


## 11. Sets in Python

Python has a built-in `set` type.

### Creating sets

You can create sets with curly braces:

`{1, 2, 3}`

or with `set(...)`:

`set([1, 2, 2, 3])`

Notice that duplicates are automatically removed.


```python
A = {1, 2, 3}
B = set([1, 2, 2, 3, 3, 3])

print('A =', A)
print('B =', B)
```

### Adding and removing elements

- `add(x)` adds one element
- `remove(x)` removes one element and raises an error if it is missing
- `discard(x)` removes one element if present, but does not raise an error if missing
- `pop()` removes and returns an arbitrary element
- `clear()` removes all elements


```python
A = {1, 2, 3}
print('Start:', A)

A.add(4)
print('After add(4):', A)

A.remove(2)
print('After remove(2):', A)

A.discard(10)  # no error
print('After discard(10):', A)
```

### Set operations in Python methods

Besides operators like `|`, `&`, `-`, and `^`, Python also provides methods:

- `union()`
- `intersection()`
- `difference()`
- `symmetric_difference()`


```python
A = {1, 2, 3}
B = {3, 4, 5}

print('union:', A.union(B))
print('intersection:', A.intersection(B))
print('difference:', A.difference(B))
print('symmetric_difference:', A.symmetric_difference(B))
```

## 12. Venn diagram thinking

Even without drawing a picture, it helps to think in terms of Venn diagrams:

- $A \cup B$: everything in either circle
- $A \cap B$: only the overlap
- $A \setminus B$: the part of $A$ outside $B$
- $B \setminus A$: the part of $B$ outside $A$
- $A \triangle B$: the non-overlapping parts


## 13. Practical Python use cases for sets

Sets are useful when you want to:

- remove duplicates
- test membership efficiently
- compare collections
- find common items or missing items


```python
# Example: remove duplicates from a list
names = ['alice', 'bob', 'alice', 'carol', 'bob']
unique_names = set(names)
print('Original list:', names)
print('Unique names:', unique_names)

# Example: find common interests
alice_interests = {'math', 'python', 'music'}
bob_interests = {'python', 'sports', 'music'}
print('Common interests:', alice_interests & bob_interests)
```

## 14. Frozen sets

A normal Python `set` is mutable, which means it can be changed.

A `frozenset` is immutable, which means it cannot be changed after creation.

This is useful when you want to use a set as a dictionary key or as an element of another set.


```python
fs = frozenset([1, 2, 3])
print('frozenset:', fs)

example_dict = {fs: 'immutable set as key'}
print(example_dict)
```

## 15. Common mistakes

### Mistake 1: confusing `{}` with an empty set in Python

In Python:

- `{}` creates an empty dictionary
- `set()` creates an empty set

### Mistake 2: expecting order

A Python set is unordered. Its printed order may vary.

### Mistake 3: trying to put mutable objects inside a set

For example, lists cannot be elements of a set because they are mutable and therefore unhashable.


```python
print(type({}))
print(type(set()))

# This would fail:
# bad_set = {[1, 2], [3, 4]}

# This works:
good_set = {(1, 2), (3, 4)}
print(good_set)
```
