## 1. Cartesian coordinates on the plane

The **Cartesian coordinate system** is the most common way to describe points in a plane.

It is built from:

- an **origin** $O=(0,0)$
- a horizontal axis called the **$x$-axis**
- a vertical axis called the **$y$-axis**

These two axes are **orthogonal**, which means they meet at a right angle.

The axes divide the plane into four parts.


```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(6,6))
plt.axhline(0)
plt.axvline(0)
plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.gca().set_aspect('equal')
plt.title("The Cartesian Coordinate Plane")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Coordinates of a point

Every point in the plane is described by a pair of numbers:

$$
(x,y)
$$

where:

- $x$ is the **horizontal coordinate**
- $y$ is the **vertical coordinate**

For example, the point

$$
(2,3)
$$

means:

- move 2 units in the positive $x$ direction
- move 3 units in the positive $y$ direction


```python
point = (2, 3)

plt.figure(figsize=(6,6))
plt.axhline(0)
plt.axvline(0)

# point and guide lines
plt.scatter(*point, s=80)
plt.plot([point[0], point[0]], [0, point[1]], linestyle='--')
plt.plot([0, point[0]], [point[1], point[1]], linestyle='--')

plt.text(point[0] + 0.2, point[1] + 0.2, "(2,3)")
plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.gca().set_aspect('equal')
plt.title("Point (2,3) in the Cartesian Plane")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. The origin

The origin is the point where the two axes intersect:

$$
O=(0,0)
$$

It is the reference point for the whole coordinate system.


```python
origin = (0, 0)

plt.figure(figsize=(6,6))
plt.axhline(0)
plt.axvline(0)
plt.scatter(*origin, s=100)
plt.text(0.2, 0.2, "O = (0,0)")
plt.xlim(-3, 3)
plt.ylim(-3, 3)
plt.gca().set_aspect('equal')
plt.title("The Origin")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Quadrants

The $x$-axis and $y$-axis divide the plane into four parts called **quadrants**.

The quadrants are numbered as follows:

- **Quadrant I**: $x>0$, $y>0$
- **Quadrant II**: $x<0$, $y>0$
- **Quadrant III**: $x<0$, $y<0$
- **Quadrant IV**: $x>0$, $y<0$

Points on the axes do **not** belong to any quadrant, because one coordinate is zero.


```python
plt.figure(figsize=(7,7))
plt.axhline(0)
plt.axvline(0)

plt.text(2, 2, "I", fontsize=18)
plt.text(-3, 2, "II", fontsize=18)
plt.text(-3, -3, "III", fontsize=18)
plt.text(2, -3, "IV", fontsize=18)

plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.gca().set_aspect('equal')
plt.title("The Four Quadrants")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. Examples of points in quadrants

From your text:

- $(2,3)$ is in Quadrant I
- $(-3,1)$ is in Quadrant II
- $(-1.5,-2.5)$ is in Quadrant III
- the origin $(0,0)$ is not in any quadrant


```python
points = {
    "(2,3)": (2, 3),
    "(-3,1)": (-3, 1),
    "(-1.5,-2.5)": (-1.5, -2.5),
    "(0,0)": (0, 0),
}

plt.figure(figsize=(7,7))
plt.axhline(0)
plt.axvline(0)

for label, (x, y) in points.items():
    plt.scatter(x, y, s=80)
    plt.text(x + 0.15, y + 0.15, label)

plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.gca().set_aspect('equal')
plt.title("Example Points and Their Quadrants")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()
```

## 6. Determining the quadrant of a point

A simple rule:

- if $x>0$ and $y>0$: Quadrant I
- if $x<0$ and $y>0$: Quadrant II
- if $x<0$ and $y<0$: Quadrant III
- if $x>0$ and $y<0$: Quadrant IV
- if $x=0$ or $y=0$: on an axis, not in a quadrant


```python
def quadrant(x, y):
    if x > 0 and y > 0:
        return "Quadrant I"
    elif x < 0 and y > 0:
        return "Quadrant II"
    elif x < 0 and y < 0:
        return "Quadrant III"
    elif x > 0 and y < 0:
        return "Quadrant IV"
    else:
        return "On an axis or at the origin"

test_points = [(2,3), (-3,1), (-1.5,-2.5), (4,-2), (0,5), (0,0)]
for p in test_points:
    print(f"{p}: {quadrant(*p)}")
```

## 7. Distance from the origin

Suppose we want the distance from the origin to the point

$$
(-3,1)
$$

Using the Pythagorean theorem:

- horizontal side length = $3$
- vertical side length = $1$

So the distance is:

$$
\sqrt{3^2 + 1^2} = \sqrt{10}
$$


```python
point = (-3, 1)

plt.figure(figsize=(7,6))
plt.axhline(0)
plt.axvline(0)

# draw point
plt.scatter(*point, s=80)
plt.text(point[0] - 0.8, point[1] + 0.2, "(-3,1)")

# rectangle edges / triangle legs
plt.plot([0, point[0]], [0, 0], linestyle='--')
plt.plot([point[0], point[0]], [0, point[1]], linestyle='--')

# distance line
plt.plot([0, point[0]], [0, point[1]])

plt.xlim(-5, 2)
plt.ylim(-2, 3)
plt.gca().set_aspect('equal')
plt.title("Distance from the Origin to (-3,1)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()

distance = (point[0]**2 + point[1]**2) ** 0.5
print("Distance from origin =", distance)
print("Exact form = sqrt(10)")
```

## 8. Distance between two points

For two points

$$
(x_1,y_1) \quad \text{and} \quad (x_2,y_2)
$$

the distance between them is

$$
d = \sqrt{(x_1-x_2)^2 + (y_1-y_2)^2}
$$

This formula comes directly from the Pythagorean theorem.


```python
def distance(p1, p2):
    x1, y1 = p1
    x2, y2 = p2
    return ((x1 - x2)**2 + (y1 - y2)**2) ** 0.5

p1 = (1, 2)
p2 = (4, 6)

print("p1 =", p1)
print("p2 =", p2)
print("distance =", distance(p1, p2))
```

## 9. Illustration of the distance formula

The differences

$$
x_1-x_2
$$

and

$$
y_1-y_2
$$

give the horizontal and vertical side lengths of a right triangle.


```python
p1 = (1, 2)
p2 = (4, 6)

plt.figure(figsize=(7,7))
plt.axhline(0)
plt.axvline(0)

plt.scatter(*p1, s=80)
plt.scatter(*p2, s=80)
plt.text(p1[0] + 0.1, p1[1] + 0.1, "P1")
plt.text(p2[0] + 0.1, p2[1] + 0.1, "P2")

# right triangle legs
plt.plot([p1[0], p2[0]], [p1[1], p1[1]], linestyle='--')
plt.plot([p2[0], p2[0]], [p1[1], p2[1]], linestyle='--')

# distance segment
plt.plot([p1[0], p2[0]], [p1[1], p2[1]])

plt.xlim(-1, 7)
plt.ylim(-1, 8)
plt.gca().set_aspect('equal')
plt.title("Distance Between Two Points")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()

print("Distance =", distance(p1, p2))
```

## 10. Some example graphs in the Cartesian plane

The Cartesian plane is where we draw graphs of equations.

For example:

- the line
$$
y=x
$$

- the parabola
$$
y=x^2
$$

The line $y=x$ passes through Quadrants I and III.

The parabola $y=x^2$ passes through Quadrant II, the origin, and Quadrant I.


```python
x = np.linspace(-4, 4, 400)

plt.figure(figsize=(7,6))
plt.plot(x, x, label="y = x")
plt.plot(x, x**2, label="y = x^2")
plt.axhline(0)
plt.axvline(0)
plt.xlim(-4, 4)
plt.ylim(-4, 10)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Examples of Graphs in the Cartesian Plane")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 11. Cartesian coordinates in 3D

We can extend the idea by adding a third axis, the **$z$-axis**, orthogonal to both the $x$- and $y$-axes.

Then a point is written as

$$
(x,y,z)
$$

instead of $(x,y)$.


```python
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

# axes
ax.plot([0, 3], [0, 0], [0, 0])
ax.plot([0, 0], [0, 3], [0, 0])
ax.plot([0, 0], [0, 0], [0, 3])

point3d = (2, 1, 2.5)
ax.scatter(*point3d, s=60)
ax.text(point3d[0], point3d[1], point3d[2], " (2,1,2.5)")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.set_title("Cartesian Coordinates in 3D")
plt.show()
```

## 12. Distance in 3D

In three dimensions, the distance between

$$
(x_1,y_1,z_1)
$$

and

$$
(x_2,y_2,z_2)
$$

is

$$
d = \sqrt{(x_1-x_2)^2 + (y_1-y_2)^2 + (z_1-z_2)^2}
$$


```python
def distance_3d(p1, p2):
    x1, y1, z1 = p1
    x2, y2, z2 = p2
    return ((x1-x2)**2 + (y1-y2)**2 + (z1-z2)**2) ** 0.5

p1 = (0, 0, 0)
p2 = (2, 1, 2.5)
print("3D distance =", distance_3d(p1, p2))
```

## 13. Higher-dimensional Cartesian space

The Cartesian coordinate idea extends beyond 2D and 3D.

In an $n$-dimensional Cartesian space:

- there are $n$ orthogonal axes
- each point is described by $n$ coordinates

For example, a city database might represent each city as a point in a 4D space:

$$
(x,y,z,w)
$$

where these coordinates could mean:

- latitude
- longitude
- average temperature
- population


```python
city = {
    "name": "Example City",
    "latitude": 47.37,
    "longitude": 8.54,
    "avg_temperature": 10.5,
    "population_millions": 0.43
}

point4d = (
    city["latitude"],
    city["longitude"],
    city["avg_temperature"],
    city["population_millions"]
)

print("A city represented as a point in 4D:")
print(point4d)
```

## 14. Why Cartesian coordinates are useful

Cartesian coordinates are useful because:

- every point has a unique coordinate description
- straight lines and right angles are easy to describe
- distances are easy to compute
- they are ideal for graphs of functions

They are less natural for describing shapes like circles or spirals, where other coordinate systems can sometimes be more convenient.

The Cartesian coordinate system is the standard way to describe points in the plane.
