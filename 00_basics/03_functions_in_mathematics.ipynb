## 1. What is a function?

In mathematics, a **function** is a rule, or a kind of machine, that:

- takes an input $x$
- produces an output $y$
- and for each input there is **exactly one** output

We usually write this as:

$$
y = f(x)
$$

This is read as:

> $y$ is equal to $f$ of $x$.

Here:

- $x$ is the **input**, **argument**, or **independent variable**
- $y = f(x)$ is the **output**, **value**, or **dependent variable**

The key requirement is:

> For every allowed input $x$, there must be one and only one output.


## 2. Important warning: $f(x)$ does not mean $f \cdot x$

The expression

$$
f(x)
$$

does **not** mean “$f$ multiplied by $x$”.

Instead, it means:

> apply the function $f$ to the input $x$.

For example, if

$$
f(x) = 5x + 10,
$$

then:

$$
f(2) = 5 \cdot 2 + 10 = 20
$$

and

$$
f(7) = 5 \cdot 7 + 10 = 45.
$$


```python
def f(x):
    return 5*x + 10

for x in [2, 7]:
    print(f"f({x}) = {f(x)}")
```

## 3. Function names and variable names

The name of the function does not have to be $f$. It could be $g$, $h$, or any other symbol or word.

Also, the variable name inside the parentheses is just a placeholder.

So these all describe the same rule:

$$
f(x) = 5x + 10
$$

$$
f(r) = 5r + 10
$$

$$
f(w) = 5w + 10
$$

$$
f(Z) = 5Z + 10
$$

The variable name changes, but the function rule stays the same.


```python
def f(x):
    return 5*x + 10

def g(r):
    return 5*r + 10

def h(w):
    return 5*w + 10

values = [0, 1, 3]
for v in values:
    print(v, f(v), g(v), h(v))
```

## 4. A function can even be unnamed

Sometimes we simply write a relationship like

$$
y = x^3
$$

without explicitly naming the function.

But the structure is still the same:

- input: $x$
- rule: cube the input
- output: $y$


```python
def cube(x):
    return x**3

for x in [-2, -1, 0, 1, 2]:
    print(f"x = {x:>2} -> x^3 = {cube(x):>3}")
```

## 5. Mathematical vs programming meaning of “function”

If you have programming experience, you already know what a function is in code.

But in mathematics, a function is a more abstract object:

- it is a mapping from inputs to outputs
- it does not have to be described by code
- it does not even have to be given by a formula
- it only has to satisfy the rule: **one input, exactly one output**

So a mathematical function is broader than a programming function.


## 6. Formal definition

Suppose:

- $X$ is a set of possible inputs
- $Y$ is a set of possible outputs

A function from $X$ to $Y$ is written:

$$
f : X \to Y
$$

This means:

> $f$ maps elements of $X$ into elements of $Y$.

A formal definition says that a function is a set of ordered pairs

$$
(x,y)
$$

such that:

- $x \in X$
- $y \in Y$
- every element of $X$ appears as the first item in some pair
- no input $x$ appears with two different outputs

That last condition is the crucial one.


## 7. Domain, codomain, and range

For a function

$$
f : X \to Y,
$$

we distinguish three important sets:

### Domain
The **domain** is the set of all allowed inputs.

### Codomain
The **codomain** is the set of possible outputs we allow the function to map into.

### Range (or image)
The **range** is the set of actual outputs that the function really produces.

So:

- domain = where inputs come from
- codomain = where outputs are allowed to live
- range = outputs actually used


```python
# Example of domain, codomain, and range in Python

X = {1, 2, 3, 4}
Y = {0, 1, 4, 9, 16, 25}

def f(x):
    return x*x

range_of_f = {f(x) for x in X}

print("Domain X    =", X)
print("Codomain Y  =", Y)
print("Range f(X)  =", range_of_f)
print("Is the range a subset of the codomain?", range_of_f <= Y)
```

## 8. Illustration: domain, codomain, and range

The next cell draws a simple mapping diagram.

- left box: domain
- right box: codomain
- arrows: the mapping
- highlighted outputs used by arrows: the range


```python
import matplotlib.pyplot as plt

X = [1, 2, 3, 4]
Y = [0, 1, 4, 9, 16, 25]
mapping = {1: 1, 2: 4, 3: 9, 4: 16}

fig, ax = plt.subplots(figsize=(9, 6))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Draw boxes
ax.plot([1, 3, 3, 1, 1], [1, 1, 9, 9, 1], linewidth=2)
ax.plot([7, 9, 9, 7, 7], [1, 1, 9, 9, 1], linewidth=2)

ax.text(2, 9.4, "Domain X", ha='center', fontsize=13)
ax.text(8, 9.4, "Codomain Y", ha='center', fontsize=13)

# Positions
x_pos = {1: 7.8, 2: 6.0, 3: 4.2, 4: 2.4}
y_pos = {0: 8.3, 1: 7.1, 4: 5.9, 9: 4.7, 16: 3.5, 25: 2.3}

# Plot domain points
for value, ypos in x_pos.items():
    ax.text(2, ypos, str(value), ha='center', va='center', fontsize=12,
            bbox=dict(boxstyle="circle,pad=0.25", alpha=0.15))

# Plot codomain points, mark actual range elements differently
actual_range = set(mapping.values())
for value, ypos in y_pos.items():
    label = str(value)
    alpha = 0.35 if value in actual_range else 0.10
    ax.text(8, ypos, label, ha='center', va='center', fontsize=12,
            bbox=dict(boxstyle="circle,pad=0.25", alpha=alpha))

# Draw arrows
for x, y in mapping.items():
    ax.annotate(
        "", xy=(7.7, y_pos[y]), xytext=(2.3, x_pos[x]),
        arrowprops=dict(arrowstyle="->", lw=1.8)
    )

ax.text(8, 1.0, "Range = {1, 4, 9, 16}", ha='center', fontsize=11)
ax.set_title("Mapping Diagram: Domain, Codomain, and Range")
plt.show()
```

## 9. Ordered pairs

A function can be described as a set of ordered pairs.

For example, let

$$
A = \{1,2,3,4,5,6\}
$$

and suppose the mapping is:

- $1 \mapsto b$
- $2 \mapsto b$
- $3 \mapsto 8$
- $4 \mapsto \text{python}$
- $5 \mapsto \text{C++}$
- $6 \mapsto 1$

Then the function can be written as:

$$
\{(1,b),(2,b),(3,8),(4,\text{python}),(5,\text{C++}),(6,1)\}
$$

This is a valid function because every input appears once and only once as the first item of a pair.


```python
pairs = {
    (1, "b"),
    (2, "b"),
    (3, 8),
    (4, "python"),
    (5, "C++"),
    (6, 1),
}

print("Ordered pairs:")
for p in sorted(pairs, key=lambda t: t[0]):
    print(p)
```

## 10. Python representation of a finite function

In Python, a finite function is often represented with a dictionary.

This works well because a dictionary also associates each key with exactly one value.


```python
f_dict = {
    1: "b",
    2: "b",
    3: 8,
    4: "python",
    5: "C++",
    6: 1
}

print("Function as dictionary:", f_dict)
print("f(4) =", f_dict[4])
print("f(6) =", f_dict[6])
```

## 11. Why one input cannot have two outputs

A relation is **not** a function if one input is linked to more than one output.

For example, the relation

$$
\{(1,a), (2,b), (3,c), (3,d)\}
$$

is **not** a function, because the input $3$ has two outputs: $c$ and $d$.

That violates the definition of a function.


```python
relation = [(1, 'a'), (2, 'b'), (3, 'c'), (3, 'd')]

from collections import defaultdict

outputs_by_input = defaultdict(list)
for x, y in relation:
    outputs_by_input[x].append(y)

print("Relation:", relation)
print()

is_function = True
for x, ys in outputs_by_input.items():
    print(f"Input {x} has outputs {ys}")
    if len(set(ys)) > 1:
        is_function = False

print()
print("Does this relation define a function?", is_function)
```

## 13. Many-to-one is allowed

A common misconception is that different inputs must have different outputs.

That is false.

A function is allowed to map several inputs to the same output.

For example:

$$
f(1) = b, \qquad f(2) = b
$$

This is perfectly valid.

What is forbidden is one input having multiple outputs.


```python
f_dict = {
    1: "b",
    2: "b",
    3: 8
}

print(f_dict)
print("Different inputs can share the same output:")
print("f(1) =", f_dict[1])
print("f(2) =", f_dict[2])
```

## 14. Range is not always the same as codomain

Consider the function

$$
f : \{1,2,3,4\} \to \{0,1,4,9,16,25\}, \qquad f(x) = x^2
$$

The codomain is

$$
\{0,1,4,9,16,25\}
$$

but the actual outputs are only

$$
\{1,4,9,16\}.
$$

So the range is smaller than the codomain.


```python
X = {1, 2, 3, 4}
Y = {0, 1, 4, 9, 16, 25}
f = lambda x: x*x

range_of_f = {f(x) for x in X}

print("Codomain =", Y)
print("Range    =", range_of_f)
print("Unused codomain elements =", Y - range_of_f)
```

## 15. Mathematical notation summary

Here are the most important notations from this topic:

### Function notation
$$
y = f(x)
$$

### Mapping notation
$$
f : X \to Y
$$

### Value of a function at $x$
$$
f(x)
$$

### Ordered pair
$$
(x,y)
$$

### Membership in a set
$$
x \in X
$$

### Range / image of $X$ under $f$
$$
f(X)
$$
or more explicitly
$$
\{f(x) : x \in X\}
$$


## 16. Python example: checking whether a finite relation is a function

For finite relations represented as a list of pairs, we can write a Python helper that checks whether the relation is a function.


```python
def is_function_relation(pairs):
    seen = {}
    for x, y in pairs:
        if x in seen and seen[x] != y:
            return False
        seen[x] = y
    return True

good_relation = [(1, 'a'), (2, 'a'), (3, 'b')]
bad_relation  = [(1, 'a'), (2, 'b'), (2, 'c')]

print("good_relation:", good_relation)
print("Is function?", is_function_relation(good_relation))
print()
print("bad_relation :", bad_relation)
print("Is function?", is_function_relation(bad_relation))
```

## 17. Python example: evaluating a function on a domain

If we know the domain and the rule, we can compute the image of every element in the domain.


```python
X = {-2, -1, 0, 1, 2}

def f(x):
    return x**3

images = {x: f(x) for x in X}
range_of_f = set(images.values())

print("Images:")
for x in sorted(images):
    print(f"{x:>2} -> {images[x]:>2}")

print()
print("Range:", range_of_f)
```

## 19. Conclusion

A mathematical function is a very precise idea:

- it takes inputs from a **domain**
- it produces outputs in a **codomain**
- each input has **exactly one** output

You can describe a function:

- by a formula, such as $f(x)=5x+10$
- by a list of ordered pairs
- by a mapping diagram
- by a table
- by a rule in words

The main rule never changes:

> one input, one unique output

This is why functions are central in mathematics, computer science, physics, and many other fields.
