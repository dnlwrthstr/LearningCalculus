## 1. What is a numerical function?

A **numerical function** is a function whose **codomain** is a set of numbers.

Typical examples:

- $f : \mathbb{R} \to \mathbb{R}$
- $g : \mathbb{Z} \to \{0,1,2\}$

A numerical function does **not** necessarily need a numerical domain, but its outputs are numbers.

### Examples

1. A real-valued function of a real variable:
$$
f(x) = 3x, \qquad f : \mathbb{R} \to \mathbb{R}
$$

2. A function on the integers:
$$
g : \mathbb{Z} \to \{0,1,2\}
$$
where $g(x)$ is the remainder when $x$ is divided by 3.

3. A function from letters to numbers:
$$
a \mapsto 1, \; b \mapsto 2, \; \dots, \; z \mapsto 26
$$


```python
# Examples of numerical functions in Python

def f(x):
    return 3*x

def g(x):
    return x % 3

alphabet_position = {chr(ord('a') + i): i + 1 for i in range(26)}

print("f(4.5) =", f(4.5))
print("g(14)  =", g(14))
print("position of 'c' =", alphabet_position['c'])
print("position of 'z' =", alphabet_position['z'])
```

## 2. How to define a function

To define a numerical function

$$
y = f(x)
$$

we specify a rule that allows us to compute the output $y$ from a chosen input $x$.

Usually this rule is given by a formula.

### Examples

$$
f(x) = x^2 + 5x + 2
$$

and

$$
g(x) = \frac{x+1}{2x^2+5}
$$

with domain

$$
(-2,4]
$$

This means:

- $x > -2$
- $x \le 4$


```python
def f(x):
    return x**2 + 5*x + 2

def g(x):
    return (x + 1) / (2*x**2 + 5)

for x in [0, 1, 2]:
    print(f"x = {x}, f(x) = {f(x)}, g(x) = {g(x)}")
```

## 3. Domain restrictions

A domain restriction tells us for which input values the function is considered.

For example, if the domain is

$$
(-2,4]
$$

then:

- $-2$ is **not included**
- $4$ **is included**

Recall:

- open interval:
$$
(a,b) = \{x \mid a < x < b\}
$$

- closed interval:
$$
[a,b] = \{x \mid a \le x \le b\}
$$


```python
import numpy as np

# Sample values in (-2, 4]
x_values = np.linspace(-1.9, 4.0, 6)
print("Sample values in (-2, 4]:", x_values)
```

## 4. Piecewise-defined functions

Sometimes a function is given by different formulas on different parts of the domain.

Example:

$$
h(x)=
\begin{cases}
-x & \text{for } x<0 \\
x^2 & \text{for } 0 \le x \le 3 \\
1 & \text{for } x>3
\end{cases}
$$

This is called a **piecewise-defined function**.


```python
def h(x):
    if x < 0:
        return -x
    elif 0 <= x <= 3:
        return x**2
    else:
        return 1

test_values = [-2, -0.5, 0, 1.5, 3, 4]
for x in test_values:
    print(f"h({x}) = {h(x)}")
```

## 5. Graphs of functions

If a function is defined on real numbers or a subset of real numbers, we can draw its graph.

The graph consists of all points

$$
(x, f(x))
$$

in the plane.


```python
import matplotlib.pyplot as plt

x = np.linspace(-6, 4, 500)
y = x**2 + 5*x + 2

plt.figure(figsize=(7,5))
plt.plot(x, y)
plt.axhline(0)
plt.axvline(0)
plt.title("Graph of $f(x)=x^2+5x+2$")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.show()
```

## 6. Graph of the piecewise function

Now let us plot the piecewise function

$$
h(x)=
\begin{cases}
-x & x<0 \\
x^2 & 0 \le x \le 3 \\
1 & x>3
\end{cases}
$$


```python
x = np.linspace(-4, 6, 600)
y = np.array([h(t) for t in x])

plt.figure(figsize=(7,5))
plt.plot(x, y)
plt.axhline(0)
plt.axvline(0)
plt.title("Graph of the piecewise function $h(x)$")
plt.xlabel("x")
plt.ylabel("h(x)")
plt.show()
```

## 7. Boundedness

A function is:

- **bounded above** if there exists a number $M$ such that
  $$
  f(x) \le M
  $$
  for all $x$ in the domain.

- **bounded below** if there exists a number $m$ such that
  $$
  f(x) \ge m
  $$
  for all $x$ in the domain.

- **bounded** if it is bounded above and bounded below.


## 8. Examples of boundedness

### Example 1
$$
f_1(x)=x^2
$$

This function is bounded below by $0$, because

$$
x^2 \ge 0
$$

for all real $x$.

But it is not bounded above.

### Example 2
$$
f_2(x)=-x^2
$$

This function is bounded above by $0$, but not bounded below.

### Example 3
$$
f_3(x)=\sin\left(\frac{\pi x}{2}\right)
$$

Since sine is always between $-1$ and $1$, this function is bounded above and below.


```python
def f1(x):
    return x**2

def f2(x):
    return -x**2

def f3(x):
    return np.sin(np.pi * x / 2)

x = np.linspace(-5, 5, 600)

plt.figure(figsize=(7,5))
plt.plot(x, f1(x))
plt.title("$f_1(x)=x^2$")
plt.axhline(0)
plt.axvline(0)
plt.show()

plt.figure(figsize=(7,5))
plt.plot(x, f2(x))
plt.title("$f_2(x)=-x^2$")
plt.axhline(0)
plt.axvline(0)
plt.show()

plt.figure(figsize=(7,5))
plt.plot(x, f3(x))
plt.title(r"$f_3(x)=\sin(\pi x/2)$")
plt.axhline(0)
plt.axvline(0)
plt.show()
```

## 9. The sign function

The **sign function** is defined by

$$
\operatorname{sgn}(x)=
\begin{cases}
-1 & \text{if } x<0 \\
0 & \text{if } x=0 \\
1 & \text{if } x>0
\end{cases}
$$

It is bounded, because its outputs are only $-1$, $0$, and $1$.


```python
def sgn(x):
    if x < 0:
        return -1
    elif x == 0:
        return 0
    else:
        return 1

x = np.linspace(-5, 5, 401)
y = np.array([sgn(t) for t in x])

plt.figure(figsize=(7,5))
plt.step(x, y, where='mid')
plt.title("Sign function sgn(x)")
plt.axhline(0)
plt.axvline(0)
plt.ylim(-1.5, 1.5)
plt.show()

print("Possible output values:", {sgn(v) for v in [-3, -1, 0, 2, 7]})
```

## 10. Increasing and decreasing functions

To talk about increasing and decreasing behavior, the domain must be numerical.

A function is **increasing** if

$$
x_2 > x_1 \implies f(x_2) \ge f(x_1)
$$

A function is **strictly increasing** if

$$
x_2 > x_1 \implies f(x_2) > f(x_1)
$$

A function is **decreasing** if

$$
x_2 > x_1 \implies f(x_2) \le f(x_1)
$$

A function is **strictly decreasing** if

$$
x_2 > x_1 \implies f(x_2) < f(x_1)
$$


## 11. Examples of monotonicity

- The sign function $\operatorname{sgn}(x)$ is **increasing**, but not strictly increasing.
- The function
$$
f_4(x)=-\sqrt{x}
$$
on non-negative real numbers is **strictly decreasing**.


```python
def f4(x):
    return -np.sqrt(x)

x = np.linspace(0, 9, 400)
y = f4(x)

plt.figure(figsize=(7,5))
plt.plot(x, y)
plt.title(r"$f_4(x)=-\sqrt{x}$ on $[0,\infty)$")
plt.axhline(0)
plt.axvline(0)
plt.show()
```

## 12. Increasing and decreasing on intervals

A function may not be increasing or decreasing on its whole domain, but it may be monotone on some interval.

For example, the function

$$
f_3(x)=\sin\left(\frac{\pi x}{2}\right)
$$

is:

- increasing on $[-1,1]$
- decreasing on $[1,3]$


```python
x = np.linspace(-3, 4, 700)
y = f3(x)

plt.figure(figsize=(8,5))
plt.plot(x, y)
plt.axvspan(-1, 1, alpha=0.15)
plt.axvspan(1, 3, alpha=0.15)
plt.axhline(0)
plt.axvline(0)
plt.title(r"$f_3(x)=\sin(\pi x/2)$ with highlighted intervals")
plt.xlabel("x")
plt.ylabel("f3(x)")
plt.show()
```

## 13. Intervals

An interval is a set of real numbers lying between two fixed numbers.

### Open interval
$$
(a,b)=\{x\mid a<x<b\}
$$

### Closed interval
$$
[a,b]=\{x\mid a\le x\le b\}
$$

You may also see half-open intervals:

$$
(a,b], \qquad [a,b)
$$


```python
interval_examples = {
    "open (a,b)": "x such that a < x < b",
    "closed [a,b]": "x such that a <= x <= b",
    "half-open (a,b]": "x such that a < x <= b",
    "half-open [a,b)": "x such that a <= x < b",
}

for k, v in interval_examples.items():
    print(f"{k:16} -> {v}")
```

## 14. Symmetry

A function can have symmetry properties.

### Even function
A function is **even** if

$$
f(x)=f(-x)
$$

for all $x$ in the domain.

This means the graph is symmetric about the **y-axis**.

### Odd function
A function is **odd** if

$$
f(-x)=-f(x)
$$

for all $x$ in the domain.

This means the graph has rotational symmetry by $180^\circ$ about the origin.


## 15. Example of an even function

Consider

$$
f_5(x)=x^6-10
$$

This is even, because

$$
(-x)^6 - 10 = x^6 - 10
$$


```python
def f5(x):
    return x**6 - 10

x = np.linspace(-2, 2, 500)
y = f5(x)

plt.figure(figsize=(7,5))
plt.plot(x, y)
plt.axhline(0)
plt.axvline(0)
plt.title(r"Even function: $f_5(x)=x^6-10$")
plt.show()

for val in [0.5, 1.2]:
    print(f"f5({val}) = {f5(val)}, f5({-val}) = {f5(-val)}")
```

## 16. Example of an odd function

Consider

$$
f_6(x)=x^5
$$

This is odd, because

$$
(-x)^5 = -x^5
$$


```python
def f6(x):
    return x**5

x = np.linspace(-2, 2, 500)
y = f6(x)

plt.figure(figsize=(7,5))
plt.plot(x, y)
plt.axhline(0)
plt.axvline(0)
plt.title(r"Odd function: $f_6(x)=x^5$")
plt.show()

for val in [0.5, 1.2]:
    print(f"f6({-val}) = {f6(-val)}, -f6({val}) = {-f6(val)}")
```

## 17. A function that is neither even nor odd

Consider

$$
f_7(x)=0.5x+\cos(x)
$$

This function is neither even nor odd.


```python
def f7(x):
    return 0.5*x + np.cos(x)

x = np.linspace(-6, 6, 700)
y = f7(x)

plt.figure(figsize=(7,5))
plt.plot(x, y)
plt.axhline(0)
plt.axvline(0)
plt.title(r"Neither even nor odd: $f_7(x)=0.5x+\cos(x)$")
plt.show()

for val in [1.0, 2.0]:
    print(f"x = {val}")
    print("f7(x)   =", f7(val))
    print("f7(-x)  =", f7(-val))
    print("-f7(x)  =", -f7(val))
    print()
```

## 18. Simple numerical checks in Python

For sampled points, we can write helper functions to inspect numerical behavior approximately.

These are not proofs, but they are very useful for intuition.


```python
def sample_is_increasing(func, xs):
    ys = [func(x) for x in xs]
    return all(ys[i+1] >= ys[i] for i in range(len(ys)-1))

def sample_is_strictly_increasing(func, xs):
    ys = [func(x) for x in xs]
    return all(ys[i+1] > ys[i] for i in range(len(ys)-1))

def sample_is_decreasing(func, xs):
    ys = [func(x) for x in xs]
    return all(ys[i+1] <= ys[i] for i in range(len(ys)-1))

def sample_is_even(func, xs, tol=1e-9):
    return all(abs(func(x) - func(-x)) < tol for x in xs)

def sample_is_odd(func, xs, tol=1e-9):
    return all(abs(func(-x) + func(x)) < tol for x in xs)

xs1 = np.linspace(-1, 1, 100)
xs2 = np.linspace(0, 9, 100)

print("sgn increasing on [-1,1]?         ", sample_is_increasing(sgn, xs1))
print("sgn strictly increasing on [-1,1]?", sample_is_strictly_increasing(sgn, xs1))
print("f4 decreasing on [0,9]?           ", sample_is_decreasing(f4, xs2))
print("f5 even?                          ", sample_is_even(f5, np.linspace(-2,2,50)))
print("f6 odd?                           ", sample_is_odd(f6, np.linspace(-2,2,50)))
print("f7 even?                          ", sample_is_even(f7, np.linspace(-2,2,50)))
print("f7 odd?                           ", sample_is_odd(f7, np.linspace(-2,2,50)))
```

## 19. Real-world interpretation

Numerical functions are often used to model real processes.

For example:

- voltage as a function of current
- production loss as a function of unemployment
- distance as a function of time
- solved exercises as a function of study time

Thinking in terms of properties helps:

- **bounded** tells you whether values stay within limits
- **increasing** tells you whether outputs grow as inputs grow
- **decreasing** tells you whether outputs fall as inputs grow
- **even/odd** tells you about symmetry


## 20. Example interpretation

Suppose a function models the number of solved problems as a function of time spent studying.

Then the function is naturally:

- **bounded below** by 0
- **bounded above** by the total number of available problems
- **increasing**, because once a problem is solved, it is not unsolved


```python
# A toy example of cumulative solved problems

time_hours = np.array([0, 1, 2, 3, 4, 5, 6])
solved = np.array([0, 2, 4, 5, 7, 7, 8])

plt.figure(figsize=(7,5))
plt.plot(time_hours, solved, marker='o')
plt.title("Solved problems as a function of study time")
plt.xlabel("Time (hours)")
plt.ylabel("Solved problems")
plt.ylim(0, 10)
plt.show()

print("Bounded below by 0? ", solved.min() >= 0)
print("Bounded above by 10?", solved.max() <= 10)
print("Increasing?         ", all(solved[i+1] >= solved[i] for i in range(len(solved)-1)))
```
