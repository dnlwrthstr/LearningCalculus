## 1. Introduction

We know that integration is the inverse operation of differentiation.

We also know from the Fundamental Theorem of Calculus that definite integrals can be computed by evaluating antiderivatives at the integration bounds.

To do that in practice, we need rules for finding antiderivatives.

In this notebook, we study some of the most important and most frequently used cases:

- constant functions
- power functions
- polynomial functions
- the reciprocal function $\frac{1}{x}$


## 2. Integral of a Constant

Integration is a linear operation, so constants can be factored out:

$$
\int k\,f(x)\,dx = k\int f(x)\,dx
$$

Now consider the simplest function:

$$
f(x)=1
$$

Its antiderivative is:

$$
\int 1\,dx = x + C
$$

because

$$
\frac{d}{dx}(x+C)=1
$$

Therefore, for any constant $k$:

$$
\int k\,dx = kx + C
$$


### Examples

$$
\int 2\,dx = 2x + C
$$

$$
\int (-5)\,dx = -5x + C
$$


## 3. Integral of a Power

Recall the derivative rule:

$$
\frac{d}{dx}(x^m)=m x^{m-1}
$$

Reversing this idea gives the antiderivative rule:

$$
\int x^n\,dx = \frac{x^{n+1}}{n+1}+C
\qquad n\neq -1
$$

More generally, if there is a constant coefficient:

$$
\int kx^n\,dx = k\cdot \frac{x^{n+1}}{n+1}+C
\qquad n\neq -1
$$

This is one of the most important formulas in integration.


## 4. Why the Rule Works

Differentiate:

$$
\frac{x^{n+1}}{n+1}
$$

We get:

$$
\frac{d}{dx}\left(\frac{x^{n+1}}{n+1}\right)
=
\frac{1}{n+1}(n+1)x^n
=
x^n
$$

So it really is the antiderivative of $x^n$, provided $n\neq -1$.


## 5. Examples of Power Integrals

### Example 1

$$
\int x^2\,dx
=
\frac{x^3}{3}+C
$$

### Example 2

$$
\int x^{-5}\,dx
=
\frac{x^{-4}}{-4}+C
=
-\frac{x^{-4}}{4}+C
$$

### Example 3

$$
\int x^{1/2}\,dx
=
\frac{x^{3/2}}{3/2}+C
=
\frac{2}{3}x^{3/2}+C
$$


## 6. Integral of a Polynomial

A polynomial has the general form

$$
p(x)=a_nx^n+a_{n-1}x^{n-1}+\cdots+a_2x^2+a_1x+a_0
$$

Because integration is linear:

$$
\int (f(x)+g(x))\,dx = \int f(x)\,dx + \int g(x)\,dx
$$

So we integrate a polynomial **term by term**.

Thus:

$$
\int p(x)\,dx
=
a_n\frac{x^{n+1}}{n+1}
+
a_{n-1}\frac{x^n}{n}
+
\cdots
+
a_2\frac{x^3}{3}
+
a_1\frac{x^2}{2}
+
a_0x
+
C
$$


## 7. Worked Example: Polynomial Integral

Compute:

$$
\int (5x^2 + x - 2)\,dx
$$

Integrate each term separately:

$$
\int (5x^2 + x - 2)\,dx
=
\int 5x^2\,dx + \int x\,dx + \int (-2)\,dx
$$

$$
=
5\int x^2\,dx + \int x\,dx - 2\int 1\,dx
$$

Now apply the rules:

$$
=
5\cdot\frac{x^3}{3}
+
\frac{x^2}{2}
-
2x
+
C
$$

So the answer is

$$
\int (5x^2 + x - 2)\,dx
=
\frac{5x^3}{3}
+
\frac{x^2}{2}
-
2x
+
C
$$


## 8. Special Case: Integral of $\frac{1}{x}$

The power rule does **not** work when $n=-1$ because it would require division by zero:

$$
\frac{x^{n+1}}{n+1}
\quad\text{becomes}\quad
\frac{x^0}{0}
$$

So we need a separate rule.

We know that:

$$
\frac{d}{dx}(\ln x)=\frac{1}{x}
\qquad (x>0)
$$

Therefore:

$$
\int \frac{1}{x}\,dx = \ln x + C
$$

But to make the formula valid on both positive and negative intervals, we write:

$$
\int \frac{1}{x}\,dx = \ln|x| + C
$$

because

$$
\frac{d}{dx}(\ln|x|)=\frac{1}{x}
\qquad (x\neq 0)
$$


## 9. Summary of the Main Rules

### Constant

$$
\int k\,dx = kx + C
$$

### Power

$$
\int kx^n\,dx = k\frac{x^{n+1}}{n+1}+C
\qquad n\neq -1
$$

### Reciprocal

$$
\int \frac{1}{x}\,dx = \ln|x|+C
$$

### Polynomial

Integrate term by term.

## 10. Symbolic Verification with SymPy

We can verify the rules symbolically by differentiating the proposed antiderivatives.


```
x = sp.symbols('x', real=True)

examples = [
    ("Integral of 2", 2*x),
    ("Integral of x^2", x**3/3),
    ("Integral of x^(-5)", -x**(-4)/4),
    ("Integral of sqrt(x)", sp.Rational(2,3)*x**sp.Rational(3,2)),
    ("Integral of 5x^2 + x - 2", 5*x**3/3 + x**2/2 - 2*x)
]

for name, antiderivative in examples:
    print(name)
    print("Antiderivative:", antiderivative)
    print("Derivative check:", sp.simplify(sp.diff(antiderivative, x)))
    print("-" * 80)
```

## 11. Visualizing a Function and an Antiderivative

Let us take

$$
f(x)=2x
$$

One antiderivative is

$$
F(x)=x^2
$$

We can plot both to see the relationship between a function and one of its antiderivatives.


```
xv = np.linspace(-3, 3, 500)
f_vals = 2*xv
F_vals = xv**2

plt.figure(figsize=(8,5))
plt.plot(xv, f_vals, label='f(x)=2x')
plt.plot(xv, F_vals, label='F(x)=x^2')
plt.title("A function and one antiderivative")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

Notice that $F(x)=x^2$ is not the same as $f(x)=2x$.

Instead, $F'(x)=f(x)$.

That is the key idea of antiderivatives.


## 12. Definite Integral Example

Using the antiderivative rules, compute:

$$
\int_0^2 x^2\,dx
$$

An antiderivative is:

$$
F(x)=\frac{x^3}{3}
$$

Now evaluate:

$$
\int_0^2 x^2\,dx
=
\left[\frac{x^3}{3}\right]_0^2
=
\frac{8}{3}-0
=
\frac{8}{3}
$$


```
x = sp.symbols('x', positive=True)
result = sp.integrate(x**2, (x, 0, 2))
result
```
